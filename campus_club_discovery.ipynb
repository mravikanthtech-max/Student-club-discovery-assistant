````python
import json

def recommend_club_with_instructor_style(student_query, clubs_dataset):

    # 1. Format your dataset into text just like your instructor did with the student profile
    clubs_text = json.dumps(clubs_dataset, indent=2)

    prompt = f"""
You are the AI engine for a Campus Club Discovery Assistant.

STUDENT INTEREST / QUERY:
"{student_query}"

AVAILABLE CLUBS DATASET:
{clubs_text}

Task: Analyze the student's query and find the best matching club from the dataset.

Return ONLY valid JSON with this exact structure:

{{
    "matched_club_name": "",
    "match_reason": "",
    "suggested_subunit_or_role": ""
}}

Requirements:
- Ground your answer strictly in the provided clubs dataset.
- Keep the reason concise and clear.
"""

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b", # Your instructor's model
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.3
    )

    text = response.choices[0].message.content

    # 2. Parse the JSON just like your instructor did
    try:
        return json.loads(text)
    except Exception as e:
        print("AI returned invalid JSON:")
        print(text)
        return None
````

````python
import json
from groq import Groq

# Initialize the Groq client
client = Groq()

# 1. Define your AI Recommendation Function
def generate_club_recommendation(student_query, dataset):
    clubs_text = json.dumps(dataset, indent=2)

    prompt = f"""
You are an expert university club advisor.
Analyze the student input and recommend the single best matching club from the dataset.

STUDENT INPUT / INTERESTS:
"{student_query}"

CLUBS DATASET:
{clubs_text}

Return ONLY valid JSON with this exact structure:

{{
    "matched_club_name": "",
    "match_reason": "",
    "suggested_role_or_subunit": "",
    "key_benefits": [],
    "contact":""
}}
"""

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3
    )

    text = response.choices[0].message.content

    # Instructor's try-except JSON parsing block
    try:
        return json.loads(text)
    except Exception:
        print("AI returned invalid JSON:")
        print(text)
        return None

# 2. Interactive Manual Input in Colab
print("--- Campus Club Discovery Assistant ---")
student_interest_input = input("Enter your interests or what you want to learn: ")

print(f"\nAnalyzing match for: '{student_interest_input}'...\n")

# Call the function with your existing clubs_dataset from your notebook
recommendation = generate_club_recommendation(student_interest_input, clubs_dataset)

# Print the final result in clean, indented JSON format
print(
    json.dumps(
        recommendation,
        indent=2
    )
)
````

````python
%%writefile app.py
import base64
import html
import json
import os
import random

import streamlit as st
from groq import Groq

# ---------------------------------------------------------
# Editable site details (footer). Leave a value empty to hide it.
# ---------------------------------------------------------
COLLEGE_NAME = "Don Bosco Institute of Technology"
COLLEGE_ADDRESS = ""            # e.g. "Street, City, PIN"
COORDINATOR_NAME = ""           # e.g. "Prof. Name"
COORDINATOR_PHONE = ""          # e.g. "9876543210"
SOCIAL_LINKS = {}               # e.g. {"Instagram": "https://...", "LinkedIn": "https://..."}

LOGO_DIR = "logos"  # optional: logos/<club_id>.png  and  logos/dbit.png
GALLERY_DIR = "gallery"  # optional: gallery/<club_id>/*.jpg  (event photos for the popup)

# Optional per-club extras shown inside the details popup.
# Fill in real data only; each block appears only when it has content.
CLUB_EXTRAS = {
    # "aws_sbg": {
    #     "events": [{"date": "12 Nov", "title": "DeployFest"}],
    #     "team": [{"name": "Name", "role": "Coordinator", "photo": "logos/team/name.png"}],
    #     "quotes": [{"text": "One short line.", "by": "Member name"}],
    #     "video": "https://www.youtube.com/watch?v=...",
    # },
}


# ---------------------------------------------------------
# Page Configuration
# ---------------------------------------------------------
def _page_icon():
    for ext in ("png", "jpg", "jpeg", "webp"):
        path = os.path.join(LOGO_DIR, f"dbit.{ext}")
        if os.path.exists(path):
            try:
                from PIL import Image
                return Image.open(path)
            except Exception:
                break
    return "🎓"


st.set_page_config(
    page_title="Campus Club Discovery Assistant | DBIT",
    page_icon=_page_icon(),
    layout="wide",
)

# ---------------------------------------------------------
# Groq client (unchanged)
# ---------------------------------------------------------
@st.cache_resource
def get_groq_client():
    # Make sure your GROQ_API_KEY environment variable is set
    return Groq()

client = get_groq_client()

# ---------------------------------------------------------
# Datasets (unchanged - this exact list is what the AI sees)
# ---------------------------------------------------------
tridha_club = {
    "club_id": "tridha_cultural",
    "club_name": "Tridha — Cultural Club",
    "domain": "Culture, Performing Arts, & Digital Media",
    "description": "The campus hub for creative expression, stage performance, and event media production.",
    "sub_units": ["Tech & Media", "Fashion", "Dance", "Singing", "Theatre", "Instrumental", "Arts"],
    "key_activities": [
        "Social media & tech management for events",
        "Stage plays & street performances",
        "Live musical & dance showcases",
        "Fashion shows & art exhibitions"
    ],
    "skills_learned": [
        "Video Editing",
        "Social Media Tech Management",
        "Event Production",
        "Public Speaking",
        "Choreography",
        "Stage Design"
    ],
    "membership_requirements": "Auditions for performance and arts teams; portfolio/application for Tech & Media.",
    "contact": "9197282737"
}

aws_sbg_club = {
    "club_id": "aws_sbg",
    "club_name": "AWS Student Builder Group",
    "domain": "Cloud Computing & Infrastructure",
    "description": "A tech community dedicated to hands-on cloud skills, deployment challenges, and architecture.",
    "sub_units": ["Cloud Engineering", "Workshops & Training", "Club Operations"],
    "key_activities": [
        "Cloud architecture study jams",
        "Hands-on lab sessions on EC2, S3, & IAM",
        "DeployFest deployment challenges",
        "Cloud certification guidance"
    ],
    "roles_open": ["Club Operations", "Social Media", "Technical Team", "Onsite Support"],
    "skills_learned": [
        "Amazon Web Services (AWS)",
        "Cloud Infrastructure",
        "DevOps Fundamentals",
        "Computer Networking",
        "Server Deployment"
    ],
    "membership_requirements": "Any student who is willing to join and contribute to club while enhancing their skills in cloud but selection based on roles.",
    "contact": "5454874515"
}

open_source_club = {
    "club_id": "open_source_student_club",
    "club_name": "Open Source Student Club",
    "domain": "Software Engineering & Core Computer Science",
    "description": "A student-led initiative teaching core CSE concepts, DSA, and open-source contribution workflows.",
    "sub_units": ["DSA & Problem Solving Squad", "Core CSE Study Group", "Open Source Contribution Team"],
    "key_activities": [
        "Data Structures & Algorithms problem-solving sprints",
        "Git and GitHub open-source bootcamps",
        "Deep-dives into OS, DBMS, and Networking",
        "Coding contests and peer code reviews"
    ],
    "skills_learned": [
        "Data Structures & Algorithms",
        "Git & GitHub Workflows",
        "C++ / Java / Python",
        "Operating Systems & Linux",
        "Collaborative Software Engineering"
    ],
    "membership_requirements": "Open to all students willing to commit to regular coding sessions; basic programming knowledge.",
    "contact": "945455451"
}

infosys_springboard_club = {
    "club_id": "infosys_springboard",
    "club_name": "Infosys Springboard Campus Club",
    "domain": "Digital Upskilling, Corporate Tech & Career Readiness",
    "description": "A specialized initiative leveraging the Infosys Springboard platform to drive structured learning in programming, emerging tech, and professional soft skills.",
    "sub_units": ["Curriculum & Study Tracks", "Hackathon & Coding Squad", "Campus Outreach & Placement Prep"],
    "key_activities": [
        "Infosys Springboard certification challenges",
        "Mock interviews and aptitude training sessions",
        "Corporate readiness bootcamps"
    ],
    "roles_open": ["Campus Lead / Ambassador", "Study Track Mentor", "Technical Coordinator", "Event Promotion Team"],
    "skills_learned": [
        "Enterprise Software Development",
        "Aptitude & Problem Solving",
        "Professional Certifications",
        "Corporate Soft Skills"
    ],
    "membership_requirements": "Open to all students willing to complete structured learning modules and certification tracks on the platform.",
    "contact": "8154547112"
}

ieee_club = {
    "club_id": "ieee_student_chapter",
    "club_name": "IEEE Student Chapter",
    "domain": "Advanced Computing, Research & Engineering",
    "description": "A prestigious global technical society chapter focusing on technical paper research, engineering innovation, and professional development.",
    "sub_units": ["Technical Research & Publications", "Hardware & Embedded Systems", "Events & Conference Management", "Membership & Public Relations"],
    "key_activities": [
        "Technical paper presentation workshops",
        "Hardware prototyping and IoT bootcamps",
        "National and international coding/engineering symposiums",
        "Guest lectures by industry researchers"
    ],
    "roles_open": ["Editorial & Content Team", "Technical Event Coordinator", "Logistics & Operations", "Design Lead"],
    "skills_learned": [
        "Technical Research & Writing",
        "Embedded Systems / IoT",
        "Project Management",
        "Public Speaking & Presentation"
    ],
    "membership_requirements": "Open to all engineering students; formal registration via IEEE membership and brief interview round.",
    "contact": "8745445475"
}

ipi_club = {
    "club_id": "ipi_student_chapter",
    "club_name": "Indian Plastics Institute (IPI) Student Chapter",
    "domain": "Polymer Science, Plastics Engineering & Sustainable Materials",
    "description": "A specialized technical student chapter dedicated to polymer engineering, plastics processing technology, material testing, and sustainable packaging innovations.",
    "sub_units": ["Polymer Research & Material Testing", "Industrial Relations & Plant Visits", "Seminars & Technical Events"],
    "key_activities": [
        "Plastics processing, recycling, and molding workshops",
        "Industrial visits to manufacturing and polymer plants",
        "Technical paper presentations on eco-friendly polymers",
        "Interaction sessions with polymer industry experts"
    ],
    "roles_open": ["Technical Coordinator", "Industrial Relations Lead", "Event Manager", "Content & Publicity Lead"],
    "skills_learned": [
        "Polymer Science & Engineering",
        "Material Testing & Analysis",
        "Sustainable Packaging Design",
        "Industrial Manufacturing Awareness"
    ],
    "membership_requirements": "Open to all students interested in polymer technology, chemical/materials engineering, and sustainable manufacturing.",
    "contact": "8565432145"
}

clubs_dataset = [
    tridha_club,
    aws_sbg_club,
    open_source_club,
    infosys_springboard_club,
    ieee_club,
    ipi_club,
]

# ---------------------------------------------------------
# Visual identity (display only - never sent to the AI)
# "from"/"to" build the full-colour tile; "glow" is the light
# accent used on dark tiles; "solid" is the deeper accent used
# on white/dark panels where text must stay readable.
# ---------------------------------------------------------
CLUB_STYLE = {
    "tridha_cultural":          {"solid": "#9A6B1F", "glow": "#F2B33D", "from": "#0B0B10", "to": "#3B2A0A", "icon": "spark"},
    "aws_sbg":                  {"solid": "#8B3FE0", "glow": "#C9A7FF", "from": "#4B1D9C", "to": "#8B3FE0", "icon": "cloud"},
    "open_source_student_club": {"solid": "#E8710A", "glow": "#FFA24D", "from": "#2A0B30", "to": "#6A1F5A", "icon": "code"},
    "infosys_springboard":      {"solid": "#0F7BC0", "glow": "#8FD0FF", "from": "#0A4F8F", "to": "#1E9BE0", "icon": "award"},
    "ieee_student_chapter":     {"solid": "#B38600", "glow": "#FFD84D", "from": "#5C4300", "to": "#C08F00", "icon": "cpu"},
    "ipi_student_chapter":      {"solid": "#8B1E1A", "glow": "#FF9A92", "from": "#4E0C0A", "to": "#A52A24", "icon": "flask"},
}
DEFAULT_STYLE = {"solid": "#0F1B3D", "glow": "#9FB2E8", "from": "#0F1B3D", "to": "#24407F", "icon": "grad"}

# Category pills (display only)
CATEGORIES = {
    "Tech": {"aws_sbg", "open_source_student_club", "infosys_springboard", "ieee_student_chapter"},
    "Cultural": {"tridha_cultural"},
    "Research": {"ieee_student_chapter", "ipi_student_chapter"},
    "Career": {"infosys_springboard", "aws_sbg"},
}

# Badge text on each tile. Override any club here; otherwise it is derived
# from the dataset (clubs with open roles show "Recruiting now").
STATUS_OVERRIDE = {
    "tridha_cultural": "Auditions open",
}

QUICK_STARTS = [
    "I like coding",
    "I want to perform",
    "Placement prep",
    "Cloud and DevOps",
    "Research and papers",
    "Sustainable materials",
]

NAV_OPTIONS = ["Explore", "Find my club", "Compare"]

# ---------------------------------------------------------
# One consistent icon set (stroke icons, 24x24)
# ---------------------------------------------------------
ICONS = {
    "code": '<polyline points="16 18 22 12 16 6"/><polyline points="8 6 2 12 8 18"/>',
    "camera": '<path d="M23 19a2 2 0 0 1-2 2H3a2 2 0 0 1-2-2V8a2 2 0 0 1 2-2h4l2-3h6l2 3h4a2 2 0 0 1 2 2z"/><circle cx="12" cy="13" r="4"/>',
    "cloud": '<path d="M18 10h-1.26A8 8 0 1 0 9 20h9a5 5 0 0 0 0-10z"/>',
    "cpu": '<rect x="4" y="4" width="16" height="16" rx="2"/><rect x="9" y="9" width="6" height="6"/><path d="M9 1v3M15 1v3M9 20v3M15 20v3M20 9h3M20 14h3M1 9h3M1 14h3"/>',
    "mic": '<path d="M12 1a3 3 0 0 0-3 3v8a3 3 0 0 0 6 0V4a3 3 0 0 0-3-3z"/><path d="M19 10v2a7 7 0 0 1-14 0v-2M12 19v4M8 23h8"/>',
    "flask": '<path d="M9 3h6M10 3v6L4 20a1 1 0 0 0 .9 1.5h14.2A1 1 0 0 0 20 20l-6-11V3"/>',
    "award": '<circle cx="12" cy="8" r="7"/><polyline points="8.21 13.89 7 23 12 20 17 23 15.79 13.88"/>',
    "users": '<path d="M17 21v-2a4 4 0 0 0-4-4H5a4 4 0 0 0-4 4v2"/><circle cx="9" cy="7" r="4"/><path d="M23 21v-2a4 4 0 0 0-3-3.87M16 3.13a4 4 0 0 1 0 7.75"/>',
    "spark": '<polygon points="12 2 15.09 8.26 22 9.27 17 14.14 18.18 21.02 12 17.77 5.82 21.02 7 14.14 2 9.27 8.91 8.26 12 2"/>',
    "grad": '<path d="M22 10L12 5 2 10l10 5 10-5z"/><path d="M6 12v5c3 3 9 3 12 0v-5"/>',
    "phone": '<path d="M22 16.92v3a2 2 0 0 1-2.18 2 19.8 19.8 0 0 1-8.63-3.07 19.5 19.5 0 0 1-6-6A19.8 19.8 0 0 1 2.12 4.18 2 2 0 0 1 4.11 2h3a2 2 0 0 1 2 1.72c.13.96.36 1.9.7 2.81a2 2 0 0 1-.45 2.11L8.09 9.91a16 16 0 0 0 6 6l1.27-1.27a2 2 0 0 1 2.11-.45c.91.34 1.85.57 2.81.7A2 2 0 0 1 22 16.92z"/>',
    "pin": '<path d="M21 10c0 7-9 13-9 13S3 17 3 10a9 9 0 0 1 18 0z"/><circle cx="12" cy="10" r="3"/>',
    "calendar": '<rect x="3" y="4" width="18" height="18" rx="2"/><path d="M16 2v4M8 2v4M3 10h18"/>',
}


def icon_svg(name, size=16, stroke=2):
    inner = ICONS.get(name, ICONS["spark"])
    return (
        f'<svg class="ic" width="{size}" height="{size}" viewBox="0 0 24 24" fill="none" '
        f'stroke="currentColor" stroke-width="{stroke}" stroke-linecap="round" stroke-linejoin="round" '
        f'aria-hidden="true">{inner}</svg>'
    )


SKILL_ICON_RULES = [
    (("video", "media", "design", "stage", "art"), "camera"),
    (("code", "software", "c++", "git", "algorithm", "programming", "enterprise"), "code"),
    (("cloud", "aws", "server", "devops", "network", "linux", "operating"), "cloud"),
    (("embedded", "iot", "hardware", "system"), "cpu"),
    (("speaking", "presentation", "choreography", "event", "soft skills"), "mic"),
    (("research", "writing", "analysis", "testing", "polymer", "material", "packaging", "manufacturing"), "flask"),
    (("certification", "aptitude", "career", "professional", "corporate"), "award"),
    (("management", "project", "social", "collaborative"), "users"),
]


def skill_icon(skill):
    s = skill.lower()
    for keys, icon in SKILL_ICON_RULES:
        if any(k in s for k in keys):
            return icon
    return "spark"


# ---------------------------------------------------------
# Helpers
# ---------------------------------------------------------
def style_of(club):
    return CLUB_STYLE.get(club["club_id"], DEFAULT_STYLE)


def club_by_name(club_name):
    for c in clubs_dataset:
        if c["club_name"].strip().lower() == str(club_name).strip().lower():
            return c
    return None


def style_by_name(club_name):
    c = club_by_name(club_name)
    return style_of(c) if c else DEFAULT_STYLE


def tint(hex_color, alpha="1A"):
    return f"{hex_color}{alpha}"


def esc(text):
    return html.escape(str(text))


def html_block(raw):
    """Strip indentation/blank lines so Streamlit's markdown never treats HTML as a code block."""
    return "\n".join(line.strip() for line in raw.splitlines() if line.strip())


def short_text(text, limit=118):
    if len(text) <= limit:
        return text
    cut = text[:limit].rsplit(" ", 1)[0].rstrip(",;:.")
    return cut + "…"


def data_uri(path):
    ext = path.rsplit(".", 1)[-1].lower()
    mime = "image/svg+xml" if ext == "svg" else f"image/{'jpeg' if ext == 'jpg' else ext}"
    with open(path, "rb") as f:
        return f"data:{mime};base64,{base64.b64encode(f.read()).decode()}"


def find_logo(name):
    for ext in ("png", "jpg", "jpeg", "webp", "svg"):
        path = os.path.join(LOGO_DIR, f"{name}.{ext}")
        if os.path.exists(path):
            return data_uri(path)
    return None


def club_status(club):
    if club["club_id"] in STATUS_OVERRIDE:
        return STATUS_OVERRIDE[club["club_id"]]
    return "Recruiting now" if club.get("roles_open") else "Open to join"


def badge_html(club, size="md"):
    s = style_of(club)
    logo = find_logo(club["club_id"])
    inner = f'<img src="{logo}" alt="">' if logo else icon_svg(s["icon"], 26 if size == "md" else 32)
    return f'<div class="badge badge-{size}" style="background:{tint(s["solid"])};color:{s["solid"]};">{inner}</div>'


def chips_html(items, accent=None, limit=None, icons=False):
    items = list(items)[:limit] if limit else list(items)
    style = f' style="background:{tint(accent, "14")};color:{accent};"' if accent else ""
    out = []
    for i in items:
        ic = icon_svg(skill_icon(i), 13) if icons else ""
        out.append(f'<span class="chip"{style}>{ic}{esc(i)}</span>')
    return "".join(out)


def filter_clubs(category, query):
    result = []
    q = (query or "").strip().lower()
    for c in clubs_dataset:
        if category and category != "All" and c["club_id"] not in CATEGORIES.get(category, set()):
            continue
        if q:
            hay = " ".join(
                [c["club_name"], c["domain"], c["description"]]
                + c["sub_units"] + c["skills_learned"] + c["key_activities"]
            ).lower()
            if q not in hay:
                continue
        result.append(c)
    return result


# ---------------------------------------------------------
# Theme (light / dark)
# ---------------------------------------------------------
dark = bool(st.session_state.get("dark", False))

THEME_LIGHT = {
    "--head": "#0F1B3D", "--ink": "#1B2236", "--body": "#39415A", "--muted": "#5B6478",
    "--card": "#FFFFFF", "--line": "#E3E6EE", "--field": "#FAFBFD", "--soft": "#F1F3F9",
    "--glass": "rgba(255,255,255,0.66)", "--glass-line": "rgba(255,255,255,0.7)",
    "--bg": "linear-gradient(160deg,#E9EEFB 0%,#F7F4EC 55%,#EAF1FA 100%)",
    "--btn": "#0F1B3D", "--btn-ink": "#FFFFFF", "--btn-hover": "#1B2D63",
    "--shadow": "0 10px 30px rgba(15,27,61,0.10)",
}
THEME_DARK = {
    "--head": "#FFFFFF", "--ink": "#E8ECF8", "--body": "#B9C2DB", "--muted": "#8A94B0",
    "--card": "#151C36", "--line": "#263050", "--field": "#0F1530", "--soft": "#1B2342",
    "--glass": "rgba(21,28,54,0.62)", "--glass-line": "rgba(255,255,255,0.10)",
    "--bg": "linear-gradient(160deg,#080C1C 0%,#111A38 55%,#0B1226 100%)",
    "--btn": "#F2B33D", "--btn-ink": "#0F1B3D", "--btn-hover": "#FFC85E",
    "--shadow": "0 12px 34px rgba(0,0,0,0.45)",
}
theme_vars = "".join(f"{k}:{v};" for k, v in (THEME_DARK if dark else THEME_LIGHT).items())

per_club_css = "".join(
    f".st-key-club_{cid} {{ background-image: radial-gradient(circle at 88% -10%, rgba(255,255,255,0.26), transparent 55%), "
    f"linear-gradient(140deg, {s['from']}, {s['to']}) !important; "
    f"--glow: {s['glow']}; }}\n"
    for cid, s in CLUB_STYLE.items()
)

# ---------------------------------------------------------
# Styling
# ---------------------------------------------------------
CSS = """
<style>
@import url('https://fonts.googleapis.com/css2?family=Outfit:wght@600;700;800&family=DM+Sans:wght@400;500;600;700&display=swap');

:root { __THEME_VARS__ --gold: #F2B33D; --navy: #0F1B3D; }

.stApp { background: var(--bg); background-attachment: fixed; font-family: 'DM Sans', sans-serif; color: var(--ink); }
h1, h2, h3, .display { font-family: 'Outfit', sans-serif; letter-spacing: -0.01em; }
.ic { flex: none; vertical-align: -2px; }

/* tidy Streamlit chrome */
#MainMenu, footer { visibility: hidden; }
header[data-testid="stHeader"] { display: none; }
.block-container { max-width: 1180px; padding-top: 0.8rem; padding-bottom: 3rem; }
.stApp label, .stApp [data-testid="stWidgetLabel"] p { color: var(--ink); }
.stTextInput input, .stTextArea textarea { background: var(--field) !important; color: var(--ink) !important; border-radius: 12px !important; }
.stTextInput input::placeholder, .stTextArea textarea::placeholder { color: var(--muted) !important; opacity: .8; }
.stAlert { border-radius: 12px; }
@media (prefers-reduced-motion: reduce) { * { animation: none !important; transition: none !important; } }

/* ---------- sticky top bar ---------- */
.st-key-topbar {
    position: sticky; top: 8px; z-index: 100;
    background: var(--glass); backdrop-filter: blur(16px); -webkit-backdrop-filter: blur(16px);
    border: 1px solid var(--glass-line); border-radius: 16px; padding: 8px 16px; margin-bottom: 14px;
    box-shadow: var(--shadow);
}
.brand { display: flex; align-items: center; gap: 12px; }
.brand-mark { width: 40px; height: 40px; border-radius: 11px; background: #fff; display: flex; align-items: center; justify-content: center; overflow: hidden; color: var(--navy); flex: none; }
.brand-mark img { width: 100%; height: 100%; object-fit: contain; padding: 3px; }
.brand-name { font-family: 'Outfit', sans-serif; font-weight: 700; color: var(--head); font-size: 1.02rem; line-height: 1.15; }
.brand-sub { color: var(--muted); font-size: 0.8rem; }

/* ---------- hero ---------- */
.hero {
    background-color: var(--navy);
    background-image: radial-gradient(rgba(255,255,255,0.07) 1px, transparent 1px), radial-gradient(circle at 85% 10%, rgba(242,179,61,0.18), transparent 45%);
    background-size: 22px 22px, 100% 100%;
    border-radius: 24px; padding: 42px 52px 38px 52px; margin-bottom: 14px;
    box-shadow: var(--shadow);
}
.hero-wrap { display: flex; align-items: center; justify-content: space-between; gap: 32px; }
.hero-logos { display: grid; grid-template-columns: repeat(3, 92px); gap: 16px 18px; flex: none; padding-right: 8px; }
.hero-logos .lg {
    width: 92px; height: 92px; border-radius: 50%; background: #fff; display: flex; align-items: center; justify-content: center;
    overflow: hidden; box-shadow: 0 0 0 4px rgba(255,255,255,0.12), 0 14px 30px rgba(0,0,0,0.35); color: var(--navy);
    animation: bob 6s ease-in-out infinite;
}
.hero-logos .lg:nth-child(3n+2) { margin-top: 26px; }
.hero-logos .lg img { width: 100%; height: 100%; object-fit: contain; }
@keyframes bob { 0%, 100% { transform: translateY(0); } 50% { transform: translateY(-12px); } }
.hero-top { display: flex; align-items: center; gap: 14px; margin-bottom: 24px; }
.hero-mark { width: 58px; height: 58px; border-radius: 16px; background: #fff; display: flex; align-items: center; justify-content: center; overflow: hidden; color: var(--navy); flex: none; }
.hero-mark img { width: 100%; height: 100%; object-fit: contain; padding: 4px; }
.hero-college { color: #E6EAF7; font-weight: 700; font-size: 1.05rem; line-height: 1.2; }
.hero-tag { color: var(--gold); font-size: 0.92rem; font-weight: 500; margin-top: 2px; }
.hero-title { font-family: 'Outfit', sans-serif; font-weight: 800; color: #fff; font-size: 3.1rem; line-height: 1.05; max-width: 640px; margin: 0 0 14px 0; }
.hero-sub { color: #B9C2DB; font-size: 1.1rem; line-height: 1.55; max-width: 540px; margin: 0 0 26px 0; }
.hero-stats { display: flex; flex-wrap: wrap; gap: 12px; }
.hero-stat { border: 1px solid rgba(255,255,255,0.18); border-radius: 999px; padding: 8px 18px; color: #D6DCEE; font-size: 0.95rem; }
.hero-stat b { color: var(--gold); font-family: 'Outfit', sans-serif; font-size: 1.05rem; margin-right: 6px; }

/* ---------- section headings ---------- */
.section-title { font-family: 'Outfit', sans-serif; font-size: 1.7rem; font-weight: 700; color: var(--head); margin: 30px 0 4px 0; }
.section-sub { color: var(--muted); font-size: 1.02rem; margin-bottom: 16px; }

/* ---------- featured carousel ---------- */
.carousel { position: relative; height: 290px; border-radius: 24px; overflow: hidden; margin: 8px 0 6px 0; box-shadow: var(--shadow); }
.slide {
    position: absolute; inset: 0; display: flex; align-items: center; justify-content: space-between; gap: 24px;
    padding: 32px 48px; opacity: 0; visibility: hidden; color: #fff;
}
.slide::before { content: ""; position: absolute; inset: 0; background: radial-gradient(circle at 85% 30%, rgba(255,255,255,0.22), transparent 50%); pointer-events: none; }
.slide-text { position: relative; max-width: 560px; }
.slide-kicker { font-size: 0.9rem; font-weight: 600; opacity: .85; margin-bottom: 8px; display: flex; align-items: center; gap: 8px; }
.slide-title { font-family: 'Outfit', sans-serif; font-weight: 800; font-size: 2.1rem; line-height: 1.1; margin-bottom: 10px; }
.slide-pitch { font-size: 1.05rem; line-height: 1.5; opacity: .92; margin-bottom: 20px; }
.slide-logo { position: relative; flex: none; width: 190px; height: 190px; display: flex; align-items: center; justify-content: center; }
.slide-logo::before { content: ""; position: absolute; inset: -10px; border-radius: 50%; background: radial-gradient(circle, var(--g) 0%, transparent 68%); opacity: .55; }
.slide-logo img { position: relative; max-width: 100%; max-height: 100%; object-fit: contain; filter: drop-shadow(0 0 26px var(--g)) drop-shadow(0 10px 18px rgba(0,0,0,.35)); }
.slide-logo .ic { position: relative; color: #fff; }
.carousel:hover .slide { animation-play-state: paused; }
.join {
    display: inline-flex; align-items: center; gap: 8px; background: var(--gold); color: var(--navy) !important;
    text-decoration: none !important; font-weight: 700; padding: 12px 22px; border-radius: 12px; min-height: 44px;
}
.join:hover { filter: brightness(0.96); }

/* ---------- club tiles (full-colour) ---------- */
[class*="st-key-club_"] {
    border: none !important; border-radius: 22px !important; padding: 10px 10px 4px 10px;
    box-shadow: 0 12px 30px rgba(15,27,61,0.18); color: #fff;
    transition: transform .2s ease, box-shadow .2s ease;
    animation: fadein .45s ease both;
}
[class*="st-key-club_"]:hover { transform: translateY(-3px); box-shadow: 0 18px 40px rgba(15,27,61,0.28); }
@keyframes fadein { from { opacity: 0; transform: translateY(8px); } to { opacity: 1; transform: none; } }
__PER_CLUB_CSS__

.club-top { display: flex; justify-content: flex-end; margin-bottom: 8px; min-height: 28px; }
.pill {
    display: inline-flex; align-items: center; gap: 7px; background: rgba(255,255,255,0.18); border: 1px solid rgba(255,255,255,0.28);
    color: #fff; border-radius: 999px; padding: 4px 12px; font-size: 0.8rem; font-weight: 600; backdrop-filter: blur(6px);
}
.dot { width: 8px; height: 8px; border-radius: 50%; background: #4ADE80; box-shadow: 0 0 0 0 rgba(74,222,128,.7); animation: pulse 2s infinite; }
@keyframes pulse { 70% { box-shadow: 0 0 0 8px rgba(74,222,128,0); } 100% { box-shadow: 0 0 0 0 rgba(74,222,128,0); } }

/* flip card (hover / tap) */
.flip { perspective: 1000px; height: 200px; margin-bottom: 14px; }
.flip-in { position: relative; width: 100%; height: 100%; transition: transform .6s cubic-bezier(.2,.7,.2,1); transform-style: preserve-3d; }
.flip:hover .flip-in, .flip:active .flip-in { transform: rotateY(180deg); }
.face { position: absolute; inset: 0; border-radius: 16px; backface-visibility: hidden; -webkit-backface-visibility: hidden; display: flex; }
.face.front { align-items: center; justify-content: center; background: rgba(255,255,255,0.10); border: 1px solid rgba(255,255,255,0.18); overflow: hidden; }
.face.front::before { content: ""; position: absolute; width: 70%; height: 90%; border-radius: 50%; background: radial-gradient(circle, var(--glow) 0%, transparent 68%); opacity: .45; }
.face.front img { position: relative; max-width: 66%; max-height: 82%; object-fit: contain; filter: drop-shadow(0 0 22px var(--glow)) drop-shadow(0 8px 14px rgba(0,0,0,.35)); }
.face.front .ic { position: relative; color: #fff; }
.face.back { transform: rotateY(180deg); background: rgba(0,0,0,0.30); border: 1px solid rgba(255,255,255,0.2); flex-direction: column; justify-content: center; padding: 14px 20px; gap: 7px; backdrop-filter: blur(8px); }
.back-title { font-family: 'Outfit', sans-serif; font-weight: 700; font-size: 0.98rem; margin-bottom: 2px; }
.brow { display: flex; align-items: center; gap: 10px; font-size: 0.92rem; color: rgba(255,255,255,0.94); }
.brow .ic { color: var(--glow); }

.club-name { font-family: 'Outfit', sans-serif; font-weight: 700; font-size: 1.2rem; line-height: 1.2; color: #fff; }
.club-domain { font-size: 0.88rem; color: rgba(255,255,255,0.78); margin: 4px 0 10px 0; line-height: 1.35; }
.club-desc { color: rgba(255,255,255,0.9); font-size: 0.97rem; line-height: 1.55; margin: 0 0 12px 0; min-height: 3.1em; }
.tile .chip { background: rgba(255,255,255,0.16); color: #fff; border: 1px solid rgba(255,255,255,0.14); }
.chips-row { margin-bottom: 8px; min-height: 2.2em; }

[class*="st-key-club_"] .stButton > button {
    background: rgba(255,255,255,0.16); color: #fff; border: 1px solid rgba(255,255,255,0.34); backdrop-filter: blur(6px);
}
[class*="st-key-club_"] .stButton > button:hover { background: rgba(255,255,255,0.28); border-color: #fff; color: #fff; }

.badge { flex: none; display: flex; align-items: center; justify-content: center; overflow: hidden; }
.badge img { width: 100%; height: 100%; object-fit: contain; }
.badge-md { width: 52px; height: 52px; border-radius: 14px; }
.badge-lg { width: 64px; height: 64px; border-radius: 16px; }

.chip {
    display: inline-flex; align-items: center; gap: 6px; background: var(--soft); color: var(--body);
    border-radius: 9px; padding: 5px 11px; font-size: 0.82rem; font-weight: 500; margin: 0 6px 6px 0;
}

/* ---------- buttons ---------- */
.stButton > button {
    border-radius: 12px; font-weight: 600; font-family: 'DM Sans', sans-serif;
    border: 1px solid var(--line); padding: 0.55rem 1rem; min-height: 46px;
    background: var(--card); color: var(--ink);
}
.stButton > button:hover { border-color: var(--head); color: var(--head); }
.stButton > button[kind="primary"], .stButton > button[data-testid="stBaseButton-primary"] {
    background: var(--btn); color: var(--btn-ink); border: none; padding: 0.7rem 1.6rem; font-size: 1.02rem;
}
.stButton > button[kind="primary"]:hover, .stButton > button[data-testid="stBaseButton-primary"]:hover { background: var(--btn-hover); color: var(--btn-ink); }
.stButton > button[kind="primary"] p, .stButton > button[data-testid="stBaseButton-primary"] p { color: var(--btn-ink); }

/* ---------- AI panel (glass) ---------- */
.st-key-ai_panel {
    background: var(--glass); backdrop-filter: blur(18px); -webkit-backdrop-filter: blur(18px);
    border: 1px solid var(--glass-line) !important; border-radius: 24px !important; padding: 24px 28px 18px 28px;
    box-shadow: var(--shadow);
}
.st-key-quick .stButton > button { min-height: 38px; padding: 0.3rem 0.8rem; font-size: 0.88rem; border-radius: 999px; }

/* chat bubbles */
.msg { display: flex; gap: 12px; margin: 16px 0; align-items: flex-start; }
.msg.user { justify-content: flex-end; }
.avatar { width: 40px; height: 40px; border-radius: 50%; flex: none; display: flex; align-items: center; justify-content: center; background: var(--navy); color: var(--gold); box-shadow: 0 0 0 3px rgba(242,179,61,0.25); }
.bubble { border-radius: 18px; padding: 14px 18px; max-width: 78%; line-height: 1.55; }
.bubble.user { background: var(--btn); color: var(--btn-ink); border-bottom-right-radius: 6px; }
.bubble.ai { background: var(--card); border: 1px solid var(--line); border-top-left-radius: 6px; color: var(--body); width: 100%; max-width: 100%; }
.typing-line { height: 12px; border-radius: 8px; margin: 10px 0; }
.shimmer { background: linear-gradient(90deg, var(--soft) 25%, rgba(255,255,255,0.55) 50%, var(--soft) 75%); background-size: 200% 100%; animation: shimmer 1.3s linear infinite; }
@keyframes shimmer { from { background-position: 200% 0; } to { background-position: -200% 0; } }

/* ---------- recommendation result ---------- */
.result-top { display: flex; align-items: center; gap: 18px; margin-bottom: 12px; }
.result-label { color: var(--muted); font-weight: 600; font-size: 0.9rem; margin-bottom: 4px; }
.result-name { font-family: 'Outfit', sans-serif; font-size: 1.5rem; font-weight: 800; color: var(--head); line-height: 1.2; }
.ring { position: relative; width: 84px; height: 84px; flex: none; }
.ring svg { transform: rotate(-90deg); }
.ring .track { stroke: var(--soft); }
.ring .fill { stroke-linecap: round; animation: ringfill 1.4s ease-out both; }
@keyframes ringfill { from { stroke-dashoffset: 238.76; } }
.ring-num { position: absolute; inset: 0; display: flex; flex-direction: column; align-items: center; justify-content: center; font-family: 'Outfit', sans-serif; font-weight: 800; color: var(--head); font-size: 1.15rem; line-height: 1; }
.ring-num small { font-family: 'DM Sans', sans-serif; font-weight: 500; font-size: 0.64rem; color: var(--muted); margin-top: 3px; }
.glow-once { animation: glowonce 1.8s ease-out 1; }
@keyframes glowonce { 0% { box-shadow: 0 0 0 0 rgba(242,179,61,0.65); } 60% { box-shadow: 0 0 0 18px rgba(242,179,61,0); } 100% { box-shadow: 0 0 0 0 rgba(242,179,61,0); } }
.bubble p { color: var(--body); font-size: 1.03rem; margin: 0 0 14px 0; }
.result-role { background: var(--soft); border-radius: 12px; padding: 12px 16px; margin-bottom: 14px; color: var(--ink); }
.result-role b { color: var(--head); }
.bubble ul { margin: 0 0 16px 0; padding-left: 20px; color: var(--body); line-height: 1.7; }
.contact {
    display: inline-flex; align-items: center; gap: 8px; background: var(--gold); color: var(--navy) !important;
    text-decoration: none !important; font-weight: 700; padding: 11px 18px; border-radius: 12px; min-height: 44px;
}
.contact:hover { filter: brightness(0.96); }
.top3-title { font-family: 'Outfit', sans-serif; font-weight: 700; color: var(--head); margin: 20px 0 10px 0; }
.top3 { display: grid; grid-template-columns: repeat(3, 1fr); gap: 12px; }
.mini { border: 1px solid var(--line); background: var(--soft); border-radius: 14px; padding: 12px 14px; }
.mini.best { border-color: var(--gold); box-shadow: 0 0 0 2px rgba(242,179,61,0.35); background: var(--card); }
.mini-name { font-weight: 700; color: var(--head); font-size: 0.95rem; line-height: 1.25; margin: 8px 0 8px 0; }
.mini-bar { height: 7px; border-radius: 99px; background: var(--line); overflow: hidden; }
.mini-bar i { display: block; height: 100%; border-radius: 99px; animation: barfill 1.1s ease-out both; }
@keyframes barfill { from { width: 0; } }
.mini-pct { font-size: 0.8rem; color: var(--muted); margin-top: 6px; }
.mini-best-tag { font-size: 0.74rem; font-weight: 700; color: #8A5A00; background: var(--gold); border-radius: 99px; padding: 2px 9px; }

/* confetti (renders once) */
.confetti { position: fixed; inset: 0; pointer-events: none; z-index: 9999; overflow: hidden; }
.confetti i { position: absolute; top: -16px; width: 8px; height: 14px; border-radius: 2px; opacity: 0; animation: fall linear 1 forwards; }
@keyframes fall { 0% { opacity: 1; transform: translateY(0) rotate(0); } 100% { opacity: 0; transform: translateY(105vh) rotate(720deg); } }

/* ---------- compare ---------- */
.cmp-wrap { overflow-x: auto; }
.cmp { width: 100%; border-collapse: separate; border-spacing: 0; background: var(--card); border: 1px solid var(--line); border-radius: 16px; overflow: hidden; }
.cmp th, .cmp td { padding: 14px 16px; text-align: left; vertical-align: top; border-bottom: 1px solid var(--line); color: var(--body); font-size: 0.95rem; line-height: 1.5; }
.cmp th.rowh { width: 150px; color: var(--muted); font-weight: 600; background: var(--soft); }
.cmp tr:last-child td, .cmp tr:last-child th { border-bottom: none; }
.cmp thead th { color: #fff; font-family: 'Outfit', sans-serif; font-size: 1.05rem; }

/* ---------- dialog (glass) ---------- */
div[role="dialog"], div[data-testid="stDialog"] div[role="dialog"], div[data-testid="stModal"] div[role="dialog"] {
    background: var(--card) !important; color: var(--body) !important;
    border: 1px solid var(--glass-line); border-radius: 22px;
}
div[role="dialog"] h1, div[role="dialog"] h2, div[role="dialog"] h3,
div[role="dialog"] [data-testid="stHeading"], div[role="dialog"] [data-testid="stHeading"] * { color: var(--head) !important; }
div[role="dialog"] p, div[role="dialog"] li, div[role="dialog"] span { color: inherit; }
div[role="dialog"] button[aria-label="Close"], div[role="dialog"] button[aria-label="Close"] svg { color: var(--head) !important; }
[class*="st-key-dlg_"] { background: var(--card); color: var(--body); }
.dlg-head { display: flex; align-items: center; gap: 16px; margin-bottom: 14px; }
.dlg-title { font-family: 'Outfit', sans-serif; font-size: 1.5rem; font-weight: 800; color: var(--head); line-height: 1.2; }
.dlg-domain { color: var(--muted); margin-top: 4px; }
.dlg-label { font-family: 'Outfit', sans-serif; font-weight: 700; color: var(--head); font-size: 1.05rem; margin: 22px 0 8px 0; }
.dlg-list { margin: 0; padding-left: 20px; color: var(--body); line-height: 1.75; }
.callout { background: var(--soft); border-left: 4px solid var(--gold); border-radius: 10px; padding: 12px 16px; color: var(--ink); line-height: 1.55; }
.timeline { border-left: 2px solid var(--line); margin-left: 6px; padding-left: 16px; }
.tl-item { margin-bottom: 12px; color: var(--body); }
.tl-date { font-weight: 700; color: var(--head); }
.team { display: flex; flex-wrap: wrap; gap: 16px; }
.member { text-align: center; width: 92px; font-size: 0.85rem; color: var(--body); }
.member img, .member .ph { width: 72px; height: 72px; border-radius: 50%; object-fit: cover; margin: 0 auto 6px auto; background: var(--soft); display: flex; align-items: center; justify-content: center; }
.member b { display: block; color: var(--head); }
.quote { border-left: 3px solid var(--gold); padding: 4px 0 4px 14px; margin-bottom: 10px; color: var(--body); font-style: italic; }
.quote span { display: block; font-style: normal; color: var(--muted); font-size: 0.85rem; margin-top: 2px; }

/* ---------- footer ---------- */
.footer { margin-top: 46px; padding: 26px 30px; border-radius: 20px; background: var(--glass); backdrop-filter: blur(14px); border: 1px solid var(--glass-line); color: var(--muted); font-size: 0.92rem; line-height: 1.7; }
.footer b { color: var(--head); font-family: 'Outfit', sans-serif; font-size: 1.05rem; }
.footer a { color: var(--head); font-weight: 600; margin-right: 14px; }
.footer .row { display: flex; align-items: center; gap: 8px; }

/* ---------- mobile-first adjustments ---------- */
@media (max-width: 900px) { .hero-logos { display: none; } .st-key-topbar { position: static; } }
@media (max-width: 700px) {
    .hero { padding: 26px 22px; border-radius: 20px; }
    .hero-title { font-size: 2.1rem; }
    .hero-sub { font-size: 1rem; }
    .hero-top { margin-bottom: 16px; }
    .carousel { height: 340px; }
    .slide { flex-direction: column; align-items: flex-start; justify-content: center; padding: 24px 24px; gap: 8px; }
    .slide-logo { width: 84px; height: 84px; position: absolute; right: 18px; top: 18px; }
    .slide-title { font-size: 1.55rem; }
    .top3 { grid-template-columns: 1fr; }
    .bubble { max-width: 92%; }
    .result-top { flex-direction: column; align-items: flex-start; }
    .st-key-ai_panel { padding: 18px 16px 12px 16px; }
    .stButton > button { min-height: 50px; }
    .section-title { font-size: 1.4rem; }
}
</style>
"""

st.markdown(
    CSS.replace("__THEME_VARS__", theme_vars).replace("__PER_CLUB_CSS__", per_club_css),
    unsafe_allow_html=True,
)

# ---------------------------------------------------------
# Top bar (sticky): brand, navigation, dark mode
# ---------------------------------------------------------
college_logo = find_logo("dbit")
seal_inner = f'<img src="{college_logo}" alt="">' if college_logo else icon_svg("grad", 24)


def nav_control():
    if hasattr(st, "segmented_control"):
        v = st.segmented_control("Navigate", NAV_OPTIONS, default="Explore", key="nav", label_visibility="collapsed")
        return v or "Explore"
    return st.radio("Navigate", NAV_OPTIONS, horizontal=True, key="nav", label_visibility="collapsed")


with st.container(key="topbar"):
    try:
        tb1, tb2, tb3 = st.columns([3.1, 3.3, 1.1], vertical_alignment="center")
    except TypeError:
        tb1, tb2, tb3 = st.columns([3.1, 3.3, 1.1])
    with tb1:
        st.markdown(
            html_block(f"""
<div class="brand">
    <div class="brand-mark">{seal_inner}</div>
    <div>
        <div class="brand-name">{esc(COLLEGE_NAME)}</div>
        <div class="brand-sub">Campus Club Discovery</div>
    </div>
</div>
"""),
            unsafe_allow_html=True,
        )
    with tb2:
        page = nav_control()
    with tb3:
        st.toggle("Dark mode", key="dark")


def go_find():
    st.session_state["nav"] = "Find my club"


# ---------------------------------------------------------
# Hero
# ---------------------------------------------------------
def hero_logo_cluster():
    cells = []
    for i, c in enumerate(clubs_dataset):
        lg = find_logo(c["club_id"])
        inner = f'<img src="{lg}" alt="">' if lg else icon_svg(style_of(c)["icon"], 34)
        cells.append(f'<div class="lg" style="animation-delay:{-i * 0.9}s">{inner}</div>')
    return "".join(cells)


total_units = sum(len(c["sub_units"]) for c in clubs_dataset)
total_skills = len({s for c in clubs_dataset for s in c["skills_learned"]})

st.markdown(
    html_block(f"""
<div class="hero">
<div class="hero-wrap">
<div>
    <div class="hero-top">
        <div class="hero-mark">{seal_inner}</div>
        <div>
            <div class="hero-college">{esc(COLLEGE_NAME)}</div>
            <div class="hero-tag">Learn it. Build it. Belong.</div>
        </div>
    </div>
    <div class="hero-title">Find the club where you belong</div>
    <div class="hero-sub">Browse every student club on campus, or tell us what you enjoy and we will point you to the one that fits.</div>
    <div class="hero-stats">
        <div class="hero-stat"><b>{len(clubs_dataset)}</b>clubs</div>
        <div class="hero-stat"><b>{total_units}</b>teams you can join</div>
        <div class="hero-stat"><b>{total_skills}</b>skills you can pick up</div>
    </div>
</div>
<div class="hero-logos">{hero_logo_cluster()}</div>
</div>
</div>
"""),
    unsafe_allow_html=True,
)

if page == "Explore":
    st.button("Find my club", type="primary", on_click=go_find, key="hero_cta")


# ---------------------------------------------------------
# Club details (popup, with a fallback for older Streamlit)
# ---------------------------------------------------------
def render_club_details(club):
    with st.container(key=f"dlg_{club['club_id']}"):
        _render_club_details_inner(club)


def _render_club_details_inner(club):
    s = style_of(club)
    cid = club["club_id"]
    roles = club.get("roles_open")
    extras = CLUB_EXTRAS.get(cid, {})

    events_html = ""
    if extras.get("events"):
        items = "".join(
            f'<div class="tl-item"><span class="tl-date">{esc(e.get("date", ""))}</span> &nbsp; {esc(e.get("title", ""))}</div>'
            for e in extras["events"][:3]
        )
        events_html = f'<div class="dlg-label">Upcoming events</div><div class="timeline">{items}</div>'

    team_html = ""
    if extras.get("team"):
        members = []
        for m in extras["team"]:
            photo = m.get("photo")
            if photo and os.path.exists(photo):
                pic = f'<img src="{data_uri(photo)}" alt="">'
            else:
                pic = f'<div class="ph">{icon_svg("users", 26)}</div>'
            members.append(f'<div class="member">{pic}<b>{esc(m.get("name", ""))}</b>{esc(m.get("role", ""))}</div>')
        team_html = f'<div class="dlg-label">Meet the team</div><div class="team">{"".join(members)}</div>'

    quotes_html = ""
    if extras.get("quotes"):
        quotes_html = '<div class="dlg-label">What members say</div>' + "".join(
            f'<div class="quote">“{esc(q.get("text", ""))}”<span>{esc(q.get("by", ""))}</span></div>' for q in extras["quotes"][:3]
        )

    phone = icon_svg("phone", 16)
    st.markdown(
        html_block(f"""
<div class="dlg-head">
    {badge_html(club, "lg")}
    <div>
        <div class="dlg-title">{esc(club["club_name"])}</div>
        <div class="dlg-domain">{esc(club["domain"])}</div>
    </div>
</div>
<p style="color:var(--body);line-height:1.6;margin:0;">{esc(club["description"])}</p>
<div class="dlg-label">Teams you can join</div>
<div>{chips_html(club["sub_units"], s["solid"])}</div>
<div class="dlg-label">What the club does</div>
<ul class="dlg-list">{"".join(f"<li>{esc(a)}</li>" for a in club["key_activities"])}</ul>
<div class="dlg-label">Skills you will pick up</div>
<div>{chips_html(club["skills_learned"], icons=True)}</div>
{f'<div class="dlg-label">Open roles</div><div>{chips_html(roles)}</div>' if roles else ""}
<div class="dlg-label">How to join</div>
<div class="callout">{esc(club["membership_requirements"])}</div>
{events_html}
{team_html}
{quotes_html}
"""),
        unsafe_allow_html=True,
    )

    gallery = os.path.join(GALLERY_DIR, cid)
    if os.path.isdir(gallery):
        pics = sorted(
            os.path.join(gallery, f) for f in os.listdir(gallery)
            if f.lower().endswith((".png", ".jpg", ".jpeg", ".webp"))
        )
        if pics:
            st.markdown('<div class="dlg-label">From past events</div>', unsafe_allow_html=True)
            st.image(pics[:6], width=150)

    if extras.get("video"):
        st.markdown('<div class="dlg-label">Watch</div>', unsafe_allow_html=True)
        st.video(extras["video"])

    st.markdown(
        f'<div style="margin-top:20px;"><a class="contact" href="tel:{esc(club["contact"])}">{phone}Call {esc(club["contact"])}</a></div>',
        unsafe_allow_html=True,
    )


HAS_DIALOG = hasattr(st, "dialog")

if HAS_DIALOG:
    @st.dialog("Club details", width="large")
    def club_dialog(club):
        render_club_details(club)


# ---------------------------------------------------------
# Explore: carousel, filters, tiles
# ---------------------------------------------------------
def carousel_html():
    n = len(clubs_dataset)
    slot = 100.0 / n
    dur = n * 6
    keyframes = (
        f"@keyframes slideshow {{0%{{opacity:0;visibility:visible}} {slot*0.12:.2f}%{{opacity:1;visibility:visible}} "
        f"{slot*0.88:.2f}%{{opacity:1;visibility:visible}} {slot:.2f}%{{opacity:0;visibility:hidden}} "
        f"100%{{opacity:0;visibility:hidden}}}}"
    )
    slides = []
    for i, c in enumerate(clubs_dataset):
        s = style_of(c)
        logo = find_logo(c["club_id"])
        art = f'<img src="{logo}" alt="">' if logo else icon_svg(s["icon"], 96, 1.5)
        slides.append(f"""
<div class="slide" style="background:linear-gradient(135deg,{s["from"]},{s["to"]});--g:{s["glow"]};animation:slideshow {dur}s linear {i * 6}s infinite;">
    <div class="slide-text">
        <div class="slide-kicker">{icon_svg("spark", 15)} Featured club</div>
        <div class="slide-title">{esc(c["club_name"])}</div>
        <div class="slide-pitch">{esc(short_text(c["description"], 110))}</div>
        <a class="join" href="tel:{esc(c["contact"])}">{icon_svg("phone", 16)}Join</a>
    </div>
    <div class="slide-logo">{art}</div>
</div>""")
    return html_block(f"<style>{keyframes}</style><div class=\"carousel\">{''.join(slides)}</div>")


def tile_html(club, wide):
    s = style_of(club)
    logo = find_logo(club["club_id"])
    front = f'<img src="{logo}" alt="">' if logo else icon_svg(s["icon"], 72, 1.5)
    back_rows = "".join(
        f'<div class="brow">{icon_svg(skill_icon(k), 15)}<span>{esc(k)}</span></div>'
        for k in club["skills_learned"][:5]
    )
    return html_block(f"""
<div class="tile">
<div class="club-top"><span class="pill"><i class="dot"></i>{esc(club_status(club))}</span></div>
<div class="flip"><div class="flip-in">
    <div class="face front">{front}</div>
    <div class="face back"><div class="back-title">Top skills</div>{back_rows}</div>
</div></div>
<div class="club-name">{esc(club["club_name"])}</div>
<div class="club-domain">{esc(club["domain"])}</div>
<p class="club-desc">{esc(short_text(club["description"], 150 if wide else 118))}</p>
<div class="chips-row">{chips_html(club["skills_learned"], limit=5 if wide else 3, icons=True)}</div>
</div>
""")


def choose_category():
    options = ["All"] + list(CATEGORIES.keys())
    if hasattr(st, "pills"):
        v = st.pills("Category", options, default="All", key="cat", label_visibility="collapsed")
        return v or "All"
    return st.radio("Category", options, horizontal=True, key="cat", label_visibility="collapsed")


def render_tile(club, wide):
    with st.container(border=True, key=f"club_{club['club_id']}"):
        st.markdown(tile_html(club, wide), unsafe_allow_html=True)
        if HAS_DIALOG:
            if st.button("View details", key=f"btn_{club['club_id']}", use_container_width=True):
                club_dialog(club)
        else:
            with st.expander("View details"):
                render_club_details(club)


def render_explore():
    st.markdown(carousel_html(), unsafe_allow_html=True)

    st.markdown('<div class="section-title">Explore campus clubs</div>', unsafe_allow_html=True)
    st.markdown(
        '<div class="section-sub">Filter by category or search a skill. Hover a tile to see its top skills, then open it for the full details.</div>',
        unsafe_allow_html=True,
    )

    category = choose_category()
    query = st.text_input(
        "Search clubs",
        placeholder="Search by club, skill or activity, e.g. AWS, dance, research",
        key="club_search",
        label_visibility="collapsed",
    )

    visible = filter_clubs(category, query)
    if not visible:
        st.info("No club matches that search. Clear the filter or try a different word.")
        return

    # Alternating wide / narrow rows give the grid some rhythm
    patterns = [(2, 1), (1, 2), (1, 1)]
    row = 0
    for i in range(0, len(visible), 2):
        chunk = visible[i:i + 2]
        pat = patterns[row % len(patterns)]
        row += 1
        if len(chunk) == 1:
            cols = st.columns([2, 1], gap="medium")
            with cols[0]:
                render_tile(chunk[0], wide=True)
        else:
            cols = st.columns(list(pat), gap="medium")
            for col, club, w in zip(cols, chunk, pat):
                with col:
                    render_tile(club, wide=(w == 2))


# ---------------------------------------------------------
# Core AI Recommendation Logic (unchanged)
# ---------------------------------------------------------
def generate_club_recommendation(student_query, dataset):
    clubs_text = json.dumps(dataset, indent=2)

    prompt = f"""
You are an expert university club advisor.
Analyze the student input and recommend the single best matching club from the dataset.

STUDENT INPUT / INTERESTS:
"{student_query}"

CLUBS DATASET:
{clubs_text}

Return ONLY valid JSON with this exact structure:

{{
    "matched_club_name": "",
    "match_reason": "",
    "suggested_role_or_subunit": "",
    "key_benefits": [],
    "contact": ""
}}
"""

    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.3
        )

        text = response.choices[0].message.content
        return json.loads(text)
    except Exception as e:
        return {"error": str(e)}


# ---------------------------------------------------------
# Extra call used only for the match meter and top-3 cards.
# It never changes which club is recommended above.
# ---------------------------------------------------------
def generate_match_scores(student_query, dataset):
    names = [c["club_name"] for c in dataset]
    prompt = f"""
Rate how well each club fits this student, from 0 to 100.

STUDENT INPUT / INTERESTS:
"{student_query}"

CLUBS DATASET:
{json.dumps(dataset, indent=2)}

Return ONLY valid JSON with this exact structure, using these exact club names {json.dumps(names)}:

{{
    "scores": [{{"club_name": "", "percent": 0}}]
}}
"""
    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.2
        )
        text = response.choices[0].message.content.strip()
        text = text.replace("```json", "").replace("```", "").strip()
        data = json.loads(text)
        scores = {}
        for item in data.get("scores", []):
            c = club_by_name(item.get("club_name", ""))
            if c:
                scores[c["club_id"]] = max(0, min(100, int(round(float(item.get("percent", 0))))))
        return scores
    except Exception:
        return {}


# ---------------------------------------------------------
# AI panel pieces
# ---------------------------------------------------------
AVATAR = f'<div class="avatar">{icon_svg("spark", 20)}</div>'


def user_bubble(text):
    return f'<div class="msg user"><div class="bubble user">{esc(text)}</div></div>'


def skeleton_html(query):
    return html_block(f"""
{user_bubble(query)}
<div class="msg">{AVATAR}<div class="bubble ai">
<div class="typing-line shimmer" style="width:42%"></div>
<div class="typing-line shimmer" style="width:92%"></div>
<div class="typing-line shimmer" style="width:78%"></div>
<div class="typing-line shimmer" style="width:55%"></div>
</div></div>
""")


def confetti_html():
    colors = ["#F2B33D", "#8B3FE0", "#0F7BC0", "#E8710A", "#4ADE80", "#FF6B6B"]
    pieces = []
    for _ in range(36):
        pieces.append(
            f'<i style="left:{random.uniform(2, 98):.1f}%;background:{random.choice(colors)};'
            f'animation-delay:{random.uniform(0, 0.7):.2f}s;animation-duration:{random.uniform(2.0, 3.4):.2f}s;"></i>'
        )
    return f'<div class="confetti">{"".join(pieces)}</div>'


def ring_html(percent, color):
    c = 238.76
    offset = c * (1 - percent / 100)
    return html_block(f"""
<div class="ring">
<svg width="84" height="84" viewBox="0 0 90 90">
<circle class="track" cx="45" cy="45" r="38" fill="none" stroke-width="9"/>
<circle class="fill" cx="45" cy="45" r="38" fill="none" stroke="{color}" stroke-width="9" stroke-dasharray="{c}" stroke-dashoffset="{offset:.2f}"/>
</svg>
<div class="ring-num">{percent}%<small>match</small></div>
</div>
""")


def top3_html(matched_club, scores):
    if not scores:
        return ""
    order = sorted(scores.items(), key=lambda kv: kv[1], reverse=True)
    ids = [matched_club["club_id"]] if matched_club else []
    ids += [cid for cid, _ in order if cid not in ids]
    ids = ids[:3]
    cards = []
    for idx, cid in enumerate(ids):
        club = next((c for c in clubs_dataset if c["club_id"] == cid), None)
        if not club or cid not in scores:
            continue
        s = style_of(club)
        pct = scores[cid]
        best = idx == 0
        tag = '<span class="mini-best-tag">Best match</span>' if best else ""
        cards.append(f"""
<div class="mini {'best' if best else ''}">
<div style="display:flex;align-items:center;justify-content:space-between;gap:8px;">{badge_html(club)}{tag}</div>
<div class="mini-name">{esc(club["club_name"])}</div>
<div class="mini-bar"><i style="width:{pct}%;background:linear-gradient(90deg,{s["from"]},{s["solid"]});"></i></div>
<div class="mini-pct">{pct}% match</div>
</div>""")
    if not cards:
        return ""
    return f'<div class="top3-title">Your top matches</div><div class="top3">{"".join(cards)}</div>'


def render_result(last):
    query = last["query"]
    recommendation = last["rec"]

    if "error" in recommendation:
        st.markdown(user_bubble(query), unsafe_allow_html=True)
        st.error(f"Could not get a recommendation. Details: {recommendation['error']}")
        return

    matched_name = recommendation.get("matched_club_name", "")
    matched_club = club_by_name(matched_name)
    ms = style_by_name(matched_name)
    scores = last.get("scores") or {}
    percent = scores.get(matched_club["club_id"]) if matched_club else None

    badge = badge_html(matched_club) if matched_club else f'<div class="badge badge-md" style="background:{tint(ms["solid"])};color:{ms["solid"]};">{icon_svg(ms["icon"], 26)}</div>'
    ring = ring_html(percent, ms["solid"]) if percent is not None else ""
    benefits = "".join(f"<li>{esc(b)}</li>" for b in recommendation.get("key_benefits", []))
    contact = str(recommendation.get("contact", ""))
    glow = "glow-once" if last.get("fresh") else ""

    if last.get("fresh"):
        st.markdown(confetti_html(), unsafe_allow_html=True)
        last["fresh"] = False

    st.markdown(
        html_block(f"""
{user_bubble(query)}
<div class="msg">{AVATAR}
<div class="bubble ai {glow}" style="border-left:5px solid {ms["solid"]};">
    <div class="result-top">
        {ring}
        {badge}
        <div>
            <div class="result-label">Your best match</div>
            <div class="result-name">{esc(matched_name)}</div>
        </div>
    </div>
    <p>{esc(recommendation.get("match_reason", ""))}</p>
    <div class="result-role"><b>Where to start:</b> {esc(recommendation.get("suggested_role_or_subunit", ""))}</div>
    {f"<ul>{benefits}</ul>" if benefits else ""}
    {f'<a class="contact" href="tel:{esc(contact)}">{icon_svg("phone", 16)}Call {esc(contact)}</a>' if contact else ""}
    {top3_html(matched_club, scores)}
</div>
</div>
"""),
        unsafe_allow_html=True,
    )


def set_interest(text):
    st.session_state["interest"] = text


def render_find():
    st.markdown('<div class="section-title" style="margin-top:8px;">Not sure where you fit?</div>', unsafe_allow_html=True)
    st.markdown(
        '<div class="section-sub">Describe what you enjoy or want to learn. We will match you with a club and suggest where to start.</div>',
        unsafe_allow_html=True,
    )

    with st.container(key="ai_panel"):
        last = st.session_state.get("last")
        if not last:
            st.markdown(
                html_block(f"""
<div class="msg">{AVATAR}<div class="bubble ai">Hi! Tell me what you enjoy or what you want to learn, and I will find the club that fits. You can also tap one of the quick starts below.</div></div>
"""),
                unsafe_allow_html=True,
            )

        with st.container(key="quick"):
            qcols = st.columns(len(QUICK_STARTS))
            for col, text in zip(qcols, QUICK_STARTS):
                with col:
                    st.button(text, key=f"qs_{text}", on_click=set_interest, args=(text,), use_container_width=True)

        student_interest_input = st.text_area(
            "What are your interests or what do you want to learn?",
            placeholder="e.g., I want to learn cloud architecture, practice AWS deployment, and help with tech operations.",
            height=120,
            key="interest",
        )

        find_clicked = st.button("Find my club", type="primary", key="find_btn")

        if find_clicked:
            if student_interest_input.strip():
                ph = st.empty()
                ph.markdown(skeleton_html(student_interest_input), unsafe_allow_html=True)
                recommendation = generate_club_recommendation(student_interest_input, clubs_dataset)
                scores = {}
                if "error" not in recommendation:
                    scores = generate_match_scores(student_interest_input, clubs_dataset)
                ph.empty()
                st.session_state["last"] = {
                    "query": student_interest_input,
                    "rec": recommendation,
                    "scores": scores,
                    "fresh": True,
                }
            else:
                st.warning("Tell us a little about your interests first.")

        last = st.session_state.get("last")
        if last:
            render_result(last)


# ---------------------------------------------------------
# Compare
# ---------------------------------------------------------
def render_compare():
    st.markdown('<div class="section-title" style="margin-top:8px;">Compare clubs</div>', unsafe_allow_html=True)
    st.markdown(
        '<div class="section-sub">Pick two or three clubs to see them side by side.</div>',
        unsafe_allow_html=True,
    )
    names = [c["club_name"] for c in clubs_dataset]
    picked = st.multiselect("Clubs to compare", names, default=names[:2], max_selections=3, label_visibility="collapsed")
    chosen = [club_by_name(n) for n in picked]
    chosen = [c for c in chosen if c]
    if len(chosen) < 2:
        st.info("Select at least two clubs to compare.")
        return

    head = "".join(
        f'<th style="background:linear-gradient(135deg,{style_of(c)["from"]},{style_of(c)["to"]});">{esc(c["club_name"])}</th>'
        for c in chosen
    )

    def row(label, fn):
        cells = "".join(f"<td>{fn(c)}</td>" for c in chosen)
        return f'<tr><th class="rowh">{label}</th>{cells}</tr>'

    rows = [
        row("Focus", lambda c: esc(c["domain"])),
        row("Teams", lambda c: chips_html(c["sub_units"], style_of(c)["solid"])),
        row("Skills", lambda c: chips_html(c["skills_learned"], icons=True)),
        row("Activities", lambda c: "<br>".join(esc(a) for a in c["key_activities"])),
        row("Open roles", lambda c: chips_html(c["roles_open"]) if c.get("roles_open") else "Not listed"),
        row("How to join", lambda c: esc(c["membership_requirements"])),
        row("Contact", lambda c: f'<a class="contact" href="tel:{esc(c["contact"])}">{icon_svg("phone", 15)}{esc(c["contact"])}</a>'),
    ]
    st.markdown(
        html_block(f'<div class="cmp-wrap"><table class="cmp"><thead><tr><th class="rowh" style="background:var(--soft);"></th>{head}</tr></thead><tbody>{"".join(rows)}</tbody></table></div>'),
        unsafe_allow_html=True,
    )


# ---------------------------------------------------------
# Page router
# ---------------------------------------------------------
if page == "Explore":
    render_explore()
elif page == "Find my club":
    render_find()
else:
    render_compare()

# ---------------------------------------------------------
# Footer
# ---------------------------------------------------------
footer_lines = [f"<b>{esc(COLLEGE_NAME)}</b>"]
if COLLEGE_ADDRESS:
    footer_lines.append(f'<div class="row">{icon_svg("pin", 15)}{esc(COLLEGE_ADDRESS)}</div>')
if COORDINATOR_PHONE:
    who = f" ({esc(COORDINATOR_NAME)})" if COORDINATOR_NAME else ""
    footer_lines.append(
        f'<div class="row">{icon_svg("phone", 15)}Contact the coordinator{who}: <a href="tel:{esc(COORDINATOR_PHONE)}" style="margin-left:4px;">{esc(COORDINATOR_PHONE)}</a></div>'
    )
if SOCIAL_LINKS:
    links = "".join(f'<a href="{esc(url)}" target="_blank" rel="noopener">{esc(name)}</a>' for name, url in SOCIAL_LINKS.items())
    footer_lines.append(f"<div>{links}</div>")
footer_lines.append("Campus Club Discovery Assistant")

st.markdown(f'<div class="footer">{"".join(l if l.startswith("<div") else f"<div>{l}</div>" for l in footer_lines)}</div>', unsafe_allow_html=True)
````

````python
%%writefile app,.py
import base64
import html
import json
import os
import random

import streamlit as st
from groq import Groq

# ---------------------------------------------------------
# Editable site details (footer). Leave a value empty to hide it.
# ---------------------------------------------------------
COLLEGE_NAME = "Don Bosco Institute of Technology"
COLLEGE_ADDRESS = ""            # e.g. "Street, City, PIN"
COORDINATOR_NAME = ""           # e.g. "Prof. Name"
COORDINATOR_PHONE = ""          # e.g. "9876543210"
SOCIAL_LINKS = {}               # e.g. {"Instagram": "https://...", "LinkedIn": "https://..."}

LOGO_DIR = "logos"  # optional: logos/<club_id>.png  and  logos/dbit.png
GALLERY_DIR = "gallery"  # optional: gallery/<club_id>/*.jpg  (event photos for the popup)

# Optional per-club extras shown inside the details popup.
# Fill in real data only; each block appears only when it has content.
CLUB_EXTRAS = {
    # "aws_sbg": {
    #     "events": [{"date": "12 Nov", "title": "DeployFest"}],
    #     "team": [{"name": "Name", "role": "Coordinator", "photo": "logos/team/name.png"}],
    #     "quotes": [{"text": "One short line.", "by": "Member name"}],
    #     "video": "https://www.youtube.com/watch?v=...",
    # },
}


# ---------------------------------------------------------
# Page Configuration
# ---------------------------------------------------------
def _page_icon():
    for ext in ("png", "jpg", "jpeg", "webp"):
        path = os.path.join(LOGO_DIR, f"dbit.{ext}")
        if os.path.exists(path):
            try:
                from PIL import Image
                return Image.open(path)
            except Exception:
                break
    return "🎓"


st.set_page_config(
    page_title="Campus Club Discovery Assistant | DBIT",
    page_icon=_page_icon(),
    layout="wide",
)

# ---------------------------------------------------------
# Groq client (unchanged)
# ---------------------------------------------------------
@st.cache_resource
def get_groq_client():
    # Make sure your GROQ_API_KEY environment variable is set
    return Groq()

client = get_groq_client()

# ---------------------------------------------------------
# Datasets (unchanged - this exact list is what the AI sees)
# ---------------------------------------------------------
tridha_club = {
    "club_id": "tridha_cultural",
    "club_name": "Tridha — Cultural Club",
    "domain": "Culture, Performing Arts, & Digital Media",
    "description": "The campus hub for creative expression, stage performance, and event media production.",
    "sub_units": ["Tech & Media", "Fashion", "Dance", "Singing", "Theatre", "Instrumental", "Arts"],
    "key_activities": [
        "Social media & tech management for events",
        "Stage plays & street performances",
        "Live musical & dance showcases",
        "Fashion shows & art exhibitions"
    ],
    "skills_learned": [
        "Video Editing",
        "Social Media Tech Management",
        "Event Production",
        "Public Speaking",
        "Choreography",
        "Stage Design"
    ],
    "membership_requirements": "Auditions for performance and arts teams; portfolio/application for Tech & Media.",
    "contact": "9197282737"
}

aws_sbg_club = {
    "club_id": "aws_sbg",
    "club_name": "AWS Student Builder Group",
    "domain": "Cloud Computing & Infrastructure",
    "description": "A tech community dedicated to hands-on cloud skills, deployment challenges, and architecture.",
    "sub_units": ["Cloud Engineering", "Workshops & Training", "Club Operations"],
    "key_activities": [
        "Cloud architecture study jams",
        "Hands-on lab sessions on EC2, S3, & IAM",
        "DeployFest deployment challenges",
        "Cloud certification guidance"
    ],
    "roles_open": ["Club Operations", "Social Media", "Technical Team", "Onsite Support"],
    "skills_learned": [
        "Amazon Web Services (AWS)",
        "Cloud Infrastructure",
        "DevOps Fundamentals",
        "Computer Networking",
        "Server Deployment"
    ],
    "membership_requirements": "Any student who is willing to join and contribute to club while enhancing their skills in cloud but selection based on roles.",
    "contact": "5454874515"
}

open_source_club = {
    "club_id": "open_source_student_club",
    "club_name": "Open Source Student Club",
    "domain": "Software Engineering & Core Computer Science",
    "description": "A student-led initiative teaching core CSE concepts, DSA, and open-source contribution workflows.",
    "sub_units": ["DSA & Problem Solving Squad", "Core CSE Study Group", "Open Source Contribution Team"],
    "key_activities": [
        "Data Structures & Algorithms problem-solving sprints",
        "Git and GitHub open-source bootcamps",
        "Deep-dives into OS, DBMS, and Networking",
        "Coding contests and peer code reviews"
    ],
    "skills_learned": [
        "Data Structures & Algorithms",
        "Git & GitHub Workflows",
        "C++ / Java / Python",
        "Operating Systems & Linux",
        "Collaborative Software Engineering"
    ],
    "membership_requirements": "Open to all students willing to commit to regular coding sessions; basic programming knowledge.",
    "contact": "945455451"
}

infosys_springboard_club = {
    "club_id": "infosys_springboard",
    "club_name": "Infosys Springboard Campus Club",
    "domain": "Digital Upskilling, Corporate Tech & Career Readiness",
    "description": "A specialized initiative leveraging the Infosys Springboard platform to drive structured learning in programming, emerging tech, and professional soft skills.",
    "sub_units": ["Curriculum & Study Tracks", "Hackathon & Coding Squad", "Campus Outreach & Placement Prep"],
    "key_activities": [
        "Infosys Springboard certification challenges",
        "Mock interviews and aptitude training sessions",
        "Corporate readiness bootcamps"
    ],
    "roles_open": ["Campus Lead / Ambassador", "Study Track Mentor", "Technical Coordinator", "Event Promotion Team"],
    "skills_learned": [
        "Enterprise Software Development",
        "Aptitude & Problem Solving",
        "Professional Certifications",
        "Corporate Soft Skills"
    ],
    "membership_requirements": "Open to all students willing to complete structured learning modules and certification tracks on the platform.",
    "contact": "8154547112"
}

ieee_club = {
    "club_id": "ieee_student_chapter",
    "club_name": "IEEE Student Chapter",
    "domain": "Advanced Computing, Research & Engineering",
    "description": "A prestigious global technical society chapter focusing on technical paper research, engineering innovation, and professional development.",
    "sub_units": ["Technical Research & Publications", "Hardware & Embedded Systems", "Events & Conference Management", "Membership & Public Relations"],
    "key_activities": [
        "Technical paper presentation workshops",
        "Hardware prototyping and IoT bootcamps",
        "National and international coding/engineering symposiums",
        "Guest lectures by industry researchers"
    ],
    "roles_open": ["Editorial & Content Team", "Technical Event Coordinator", "Logistics & Operations", "Design Lead"],
    "skills_learned": [
        "Technical Research & Writing",
        "Embedded Systems / IoT",
        "Project Management",
        "Public Speaking & Presentation"
    ],
    "membership_requirements": "Open to all engineering students; formal registration via IEEE membership and brief interview round.",
    "contact": "8745445475"
}

ipi_club = {
    "club_id": "ipi_student_chapter",
    "club_name": "Indian Plastics Institute (IPI) Student Chapter",
    "domain": "Polymer Science, Plastics Engineering & Sustainable Materials",
    "description": "A specialized technical student chapter dedicated to polymer engineering, plastics processing technology, material testing, and sustainable packaging innovations.",
    "sub_units": ["Polymer Research & Material Testing", "Industrial Relations & Plant Visits", "Seminars & Technical Events"],
    "key_activities": [
        "Plastics processing, recycling, and molding workshops",
        "Industrial visits to manufacturing and polymer plants",
        "Technical paper presentations on eco-friendly polymers",
        "Interaction sessions with polymer industry experts"
    ],
    "roles_open": ["Technical Coordinator", "Industrial Relations Lead", "Event Manager", "Content & Publicity Lead"],
    "skills_learned": [
        "Polymer Science & Engineering",
        "Material Testing & Analysis",
        "Sustainable Packaging Design",
        "Industrial Manufacturing Awareness"
    ],
    "membership_requirements": "Open to all students interested in polymer technology, chemical/materials engineering, and sustainable manufacturing.",
    "contact": "8565432145"
}

clubs_dataset = [
    tridha_club,
    aws_sbg_club,
    open_source_club,
    infosys_springboard_club,
    ieee_club,
    ipi_club,
]

# ---------------------------------------------------------
# Visual identity (display only - never sent to the AI)
# "from"/"to" build the full-colour tile; "glow" is the light
# accent used on dark tiles; "solid" is the deeper accent used
# on white/dark panels where text must stay readable.
# ---------------------------------------------------------
CLUB_STYLE = {
    "tridha_cultural":          {"solid": "#9A6B1F", "glow": "#F2B33D", "from": "#0B0B10", "to": "#3B2A0A", "icon": "spark"},
    "aws_sbg":                  {"solid": "#8B3FE0", "glow": "#C9A7FF", "from": "#4B1D9C", "to": "#8B3FE0", "icon": "cloud"},
    "open_source_student_club": {"solid": "#E8710A", "glow": "#FFA24D", "from": "#2A0B30", "to": "#6A1F5A", "icon": "code"},
    "infosys_springboard":      {"solid": "#0F7BC0", "glow": "#8FD0FF", "from": "#0A4F8F", "to": "#1E9BE0", "icon": "award"},
    "ieee_student_chapter":     {"solid": "#B38600", "glow": "#FFD84D", "from": "#5C4300", "to": "#C08F00", "icon": "cpu"},
    "ipi_student_chapter":      {"solid": "#8B1E1A", "glow": "#FF9A92", "from": "#4E0C0A", "to": "#A52A24", "icon": "flask"},
}
DEFAULT_STYLE = {"solid": "#0F1B3D", "glow": "#9FB2E8", "from": "#0F1B3D", "to": "#24407F", "icon": "grad"}

# Category pills (display only)
CATEGORIES = {
    "Tech": {"aws_sbg", "open_source_student_club", "infosys_springboard", "ieee_student_chapter"},
    "Cultural": {"tridha_cultural"},
    "Research": {"ieee_student_chapter", "ipi_student_chapter"},
    "Career": {"infosys_springboard", "aws_sbg"},
}

# Badge text on each tile. Override any club here; otherwise it is derived
# from the dataset (clubs with open roles show "Recruiting now").
STATUS_OVERRIDE = {
    "tridha_cultural": "Auditions open",
}

QUICK_STARTS = [
    "I like coding",
    "I want to perform",
    "Placement prep",
    "Cloud and DevOps",
    "Research and papers",
    "Sustainable materials",
]

NAV_OPTIONS = ["Explore", "Find my club", "Compare"]

# ---------------------------------------------------------
# One consistent icon set (stroke icons, 24x24)
# ---------------------------------------------------------
ICONS = {
    "code": '<polyline points="16 18 22 12 16 6"/><polyline points="8 6 2 12 8 18"/>',
    "camera": '<path d="M23 19a2 2 0 0 1-2 2H3a2 2 0 0 1-2-2V8a2 2 0 0 1 2-2h4l2-3h6l2 3h4a2 2 0 0 1 2 2z"/><circle cx="12" cy="13" r="4"/>',
    "cloud": '<path d="M18 10h-1.26A8 8 0 1 0 9 20h9a5 5 0 0 0 0-10z"/>',
    "cpu": '<rect x="4" y="4" width="16" height="16" rx="2"/><rect x="9" y="9" width="6" height="6"/><path d="M9 1v3M15 1v3M9 20v3M15 20v3M20 9h3M20 14h3M1 9h3M1 14h3"/>',
    "mic": '<path d="M12 1a3 3 0 0 0-3 3v8a3 3 0 0 0 6 0V4a3 3 0 0 0-3-3z"/><path d="M19 10v2a7 7 0 0 1-14 0v-2M12 19v4M8 23h8"/>',
    "flask": '<path d="M9 3h6M10 3v6L4 20a1 1 0 0 0 .9 1.5h14.2A1 1 0 0 0 20 20l-6-11V3"/>',
    "award": '<circle cx="12" cy="8" r="7"/><polyline points="8.21 13.89 7 23 12 20 17 23 15.79 13.88"/>',
    "users": '<path d="M17 21v-2a4 4 0 0 0-4-4H5a4 4 0 0 0-4 4v2"/><circle cx="9" cy="7" r="4"/><path d="M23 21v-2a4 4 0 0 0-3-3.87M16 3.13a4 4 0 0 1 0 7.75"/>',
    "spark": '<polygon points="12 2 15.09 8.26 22 9.27 17 14.14 18.18 21.02 12 17.77 5.82 21.02 7 14.14 2 9.27 8.91 8.26 12 2"/>',
    "grad": '<path d="M22 10L12 5 2 10l10 5 10-5z"/><path d="M6 12v5c3 3 9 3 12 0v-5"/>',
    "phone": '<path d="M22 16.92v3a2 2 0 0 1-2.18 2 19.8 19.8 0 0 1-8.63-3.07 19.5 19.5 0 0 1-6-6A19.8 19.8 0 0 1 2.12 4.18 2 2 0 0 1 4.11 2h3a2 2 0 0 1 2 1.72c.13.96.36 1.9.7 2.81a2 2 0 0 1-.45 2.11L8.09 9.91a16 16 0 0 0 6 6l1.27-1.27a2 2 0 0 1 2.11-.45c.91.34 1.85.57 2.81.7A2 2 0 0 1 22 16.92z"/>',
    "pin": '<path d="M21 10c0 7-9 13-9 13S3 17 3 10a9 9 0 0 1 18 0z"/><circle cx="12" cy="10" r="3"/>',
    "calendar": '<rect x="3" y="4" width="18" height="18" rx="2"/><path d="M16 2v4M8 2v4M3 10h18"/>',
}


def icon_svg(name, size=16, stroke=2):
    inner = ICONS.get(name, ICONS["spark"])
    return (
        f'<svg class="ic" width="{size}" height="{size}" viewBox="0 0 24 24" fill="none" '
        f'stroke="currentColor" stroke-width="{stroke}" stroke-linecap="round" stroke-linejoin="round" '
        f'aria-hidden="true">{inner}</svg>'
    )


SKILL_ICON_RULES = [
    (("video", "media", "design", "stage", "art"), "camera"),
    (("code", "software", "c++", "git", "algorithm", "programming", "enterprise"), "code"),
    (("cloud", "aws", "server", "devops", "network", "linux", "operating"), "cloud"),
    (("embedded", "iot", "hardware", "system"), "cpu"),
    (("speaking", "presentation", "choreography", "event", "soft skills"), "mic"),
    (("research", "writing", "analysis", "testing", "polymer", "material", "packaging", "manufacturing"), "flask"),
    (("certification", "aptitude", "career", "professional", "corporate"), "award"),
    (("management", "project", "social", "collaborative"), "users"),
]


def skill_icon(skill):
    s = skill.lower()
    for keys, icon in SKILL_ICON_RULES:
        if any(k in s for k in keys):
            return icon
    return "spark"


# ---------------------------------------------------------
# Helpers
# ---------------------------------------------------------
def style_of(club):
    return CLUB_STYLE.get(club["club_id"], DEFAULT_STYLE)


def club_by_name(club_name):
    for c in clubs_dataset:
        if c["club_name"].strip().lower() == str(club_name).strip().lower():
            return c
    return None


def style_by_name(club_name):
    c = club_by_name(club_name)
    return style_of(c) if c else DEFAULT_STYLE


def tint(hex_color, alpha="1A"):
    return f"{hex_color}{alpha}"


def esc(text):
    return html.escape(str(text))


def html_block(raw):
    """Strip indentation/blank lines so Streamlit's markdown never treats HTML as a code block."""
    return "\n".join(line.strip() for line in raw.splitlines() if line.strip())


def short_text(text, limit=118):
    if len(text) <= limit:
        return text
    cut = text[:limit].rsplit(" ", 1)[0].rstrip(",;:.")
    return cut + "…"


def data_uri(path):
    ext = path.rsplit(".", 1)[-1].lower()
    mime = "image/svg+xml" if ext == "svg" else f"image/{'jpeg' if ext == 'jpg' else ext}"
    with open(path, "rb") as f:
        return f"data:{mime};base64,{base64.b64encode(f.read()).decode()}"


def find_logo(name):
    for ext in ("png", "jpg", "jpeg", "webp", "svg"):
        path = os.path.join(LOGO_DIR, f"{name}.{ext}")
        if os.path.exists(path):
            return data_uri(path)
    return None


def club_status(club):
    if club["club_id"] in STATUS_OVERRIDE:
        return STATUS_OVERRIDE[club["club_id"]]
    return "Recruiting now" if club.get("roles_open") else "Open to join"


def badge_html(club, size="md"):
    s = style_of(club)
    logo = find_logo(club["club_id"])
    inner = f'<img src="{logo}" alt="">' if logo else icon_svg(s["icon"], 26 if size == "md" else 32)
    return f'<div class="badge badge-{size}" style="background:{tint(s["solid"])};color:{s["solid"]};">{inner}</div>'


def chips_html(items, accent=None, limit=None, icons=False):
    items = list(items)[:limit] if limit else list(items)
    style = f' style="background:{tint(accent, "14")};color:{accent};"' if accent else ""
    out = []
    for i in items:
        ic = icon_svg(skill_icon(i), 13) if icons else ""
        out.append(f'<span class="chip"{style}>{ic}{esc(i)}</span>')
    return "".join(out)


def filter_clubs(category, query):
    result = []
    q = (query or "").strip().lower()
    for c in clubs_dataset:
        if category and category != "All" and c["club_id"] not in CATEGORIES.get(category, set()):
            continue
        if q:
            hay = " ".join(
                [c["club_name"], c["domain"], c["description"]]
                + c["sub_units"] + c["skills_learned"] + c["key_activities"]
            ).lower()
            if q not in hay:
                continue
        result.append(c)
    return result


# ---------------------------------------------------------
# Theme (light / dark)
# ---------------------------------------------------------
dark = bool(st.session_state.get("dark", False))

THEME_LIGHT = {
    "--head": "#0F1B3D", "--ink": "#1B2236", "--body": "#39415A", "--muted": "#5B6478",
    "--card": "#FFFFFF", "--line": "#E3E6EE", "--field": "#FAFBFD", "--soft": "#F1F3F9",
    "--glass": "rgba(255,255,255,0.66)", "--glass-line": "rgba(255,255,255,0.7)",
    "--bg": "linear-gradient(160deg,#E9EEFB 0%,#F7F4EC 55%,#EAF1FA 100%)",
    "--btn": "#0F1B3D", "--btn-ink": "#FFFFFF", "--btn-hover": "#1B2D63",
    "--shadow": "0 10px 30px rgba(15,27,61,0.10)",
}
THEME_DARK = {
    "--head": "#FFFFFF", "--ink": "#E8ECF8", "--body": "#B9C2DB", "--muted": "#8A94B0",
    "--card": "#151C36", "--line": "#263050", "--field": "#0F1530", "--soft": "#1B2342",
    "--glass": "rgba(21,28,54,0.62)", "--glass-line": "rgba(255,255,255,0.10)",
    "--bg": "linear-gradient(160deg,#080C1C 0%,#111A38 55%,#0B1226 100%)",
    "--btn": "#F2B33D", "--btn-ink": "#0F1B3D", "--btn-hover": "#FFC85E",
    "--shadow": "0 12px 34px rgba(0,0,0,0.45)",
}
theme_vars = "".join(f"{k}:{v};" for k, v in (THEME_DARK if dark else THEME_LIGHT).items())

per_club_css = "".join(
    f".st-key-club_{cid} {{ background-image: radial-gradient(circle at 88% -10%, rgba(255,255,255,0.26), transparent 55%), "
    f"linear-gradient(140deg, {s['from']}, {s['to']}) !important; "
    f"--glow: {s['glow']}; }}\n"
    for cid, s in CLUB_STYLE.items()
)

# ---------------------------------------------------------
# Styling
# ---------------------------------------------------------
CSS = """
<style>
@import url('https://fonts.googleapis.com/css2?family=Outfit:wght@600;700;800&family=DM+Sans:wght@400;500;600;700&display=swap');

:root { __THEME_VARS__ --gold: #F2B33D; --navy: #0F1B3D; }

.stApp { background: var(--bg); background-attachment: fixed; font-family: 'DM Sans', sans-serif; color: var(--ink); }
h1, h2, h3, .display { font-family: 'Outfit', sans-serif; letter-spacing: -0.01em; }
.ic { flex: none; vertical-align: -2px; }

/* tidy Streamlit chrome */
#MainMenu, footer { visibility: hidden; }
header[data-testid="stHeader"] { display: none; }
.block-container { max-width: 1180px; padding-top: 0.8rem; padding-bottom: 3rem; }
.stApp label, .stApp [data-testid="stWidgetLabel"] p { color: var(--ink); }
.stTextInput input, .stTextArea textarea { background: var(--field) !important; color: var(--ink) !important; border-radius: 12px !important; }
.stTextInput input::placeholder, .stTextArea textarea::placeholder { color: var(--muted) !important; opacity: .8; }
.stAlert { border-radius: 12px; }
@media (prefers-reduced-motion: reduce) { * { animation: none !important; transition: none !important; } }

/* ---------- sticky top bar ---------- */
.st-key-topbar {
    position: sticky; top: 8px; z-index: 100;
    background: var(--glass); backdrop-filter: blur(16px); -webkit-backdrop-filter: blur(16px);
    border: 1px solid var(--glass-line); border-radius: 16px; padding: 8px 16px; margin-bottom: 14px;
    box-shadow: var(--shadow);
}
.brand { display: flex; align-items: center; gap: 12px; }
.brand-mark { width: 40px; height: 40px; border-radius: 11px; background: #fff; display: flex; align-items: center; justify-content: center; overflow: hidden; color: var(--navy); flex: none; }
.brand-mark img { width: 100%; height: 100%; object-fit: contain; padding: 3px; }
.brand-name { font-family: 'Outfit', sans-serif; font-weight: 700; color: var(--head); font-size: 1.02rem; line-height: 1.15; }
.brand-sub { color: var(--muted); font-size: 0.8rem; }

/* ---------- hero ---------- */
.hero {
    background-color: var(--navy);
    background-image: radial-gradient(rgba(255,255,255,0.07) 1px, transparent 1px), radial-gradient(circle at 85% 10%, rgba(242,179,61,0.18), transparent 45%);
    background-size: 22px 22px, 100% 100%;
    border-radius: 24px; padding: 42px 52px 38px 52px; margin-bottom: 14px;
    box-shadow: var(--shadow);
}
.hero-wrap { display: flex; align-items: center; justify-content: space-between; gap: 32px; }
.hero-logos { display: grid; grid-template-columns: repeat(3, 92px); gap: 16px 18px; flex: none; padding-right: 8px; }
.hero-logos .lg {
    width: 92px; height: 92px; border-radius: 50%; background: #fff; display: flex; align-items: center; justify-content: center;
    overflow: hidden; box-shadow: 0 0 0 4px rgba(255,255,255,0.12), 0 14px 30px rgba(0,0,0,0.35); color: var(--navy);
    animation: bob 6s ease-in-out infinite;
}
.hero-logos .lg:nth-child(3n+2) { margin-top: 26px; }
.hero-logos .lg img { width: 100%; height: 100%; object-fit: contain; }
@keyframes bob { 0%, 100% { transform: translateY(0); } 50% { transform: translateY(-12px); } }
.hero-top { display: flex; align-items: center; gap: 14px; margin-bottom: 24px; }
.hero-mark { width: 58px; height: 58px; border-radius: 16px; background: #fff; display: flex; align-items: center; justify-content: center; overflow: hidden; color: var(--navy); flex: none; }
.hero-mark img { width: 100%; height: 100%; object-fit: contain; padding: 4px; }
.hero-college { color: #E6EAF7; font-weight: 700; font-size: 1.05rem; line-height: 1.2; }
.hero-tag { color: var(--gold); font-size: 0.92rem; font-weight: 500; margin-top: 2px; }
.hero-title { font-family: 'Outfit', sans-serif; font-weight: 800; color: #fff; font-size: 3.1rem; line-height: 1.05; max-width: 640px; margin: 0 0 14px 0; }
.hero-sub { color: #B9C2DB; font-size: 1.1rem; line-height: 1.55; max-width: 540px; margin: 0 0 26px 0; }
.hero-stats { display: flex; flex-wrap: wrap; gap: 12px; }
.hero-stat { border: 1px solid rgba(255,255,255,0.18); border-radius: 999px; padding: 8px 18px; color: #D6DCEE; font-size: 0.95rem; }
.hero-stat b { color: var(--gold); font-family: 'Outfit', sans-serif; font-size: 1.05rem; margin-right: 6px; }

/* ---------- section headings ---------- */
.section-title { font-family: 'Outfit', sans-serif; font-size: 1.7rem; font-weight: 700; color: var(--head); margin: 30px 0 4px 0; }
.section-sub { color: var(--muted); font-size: 1.02rem; margin-bottom: 16px; }

/* ---------- featured carousel ---------- */
.carousel { position: relative; height: 290px; border-radius: 24px; overflow: hidden; margin: 8px 0 6px 0; box-shadow: var(--shadow); }
.slide {
    position: absolute; inset: 0; display: flex; align-items: center; justify-content: space-between; gap: 24px;
    padding: 32px 48px; opacity: 0; visibility: hidden; color: #fff;
}
.slide::before { content: ""; position: absolute; inset: 0; background: radial-gradient(circle at 85% 30%, rgba(255,255,255,0.22), transparent 50%); pointer-events: none; }
.slide-text { position: relative; max-width: 560px; }
.slide-kicker { font-size: 0.9rem; font-weight: 600; opacity: .85; margin-bottom: 8px; display: flex; align-items: center; gap: 8px; }
.slide-title { font-family: 'Outfit', sans-serif; font-weight: 800; font-size: 2.1rem; line-height: 1.1; margin-bottom: 10px; }
.slide-pitch { font-size: 1.05rem; line-height: 1.5; opacity: .92; margin-bottom: 20px; }
.slide-logo { position: relative; flex: none; width: 190px; height: 190px; display: flex; align-items: center; justify-content: center; }
.slide-logo::before { content: ""; position: absolute; inset: -10px; border-radius: 50%; background: radial-gradient(circle, var(--g) 0%, transparent 68%); opacity: .55; }
.slide-logo img { position: relative; max-width: 100%; max-height: 100%; object-fit: contain; filter: drop-shadow(0 0 26px var(--g)) drop-shadow(0 10px 18px rgba(0,0,0,.35)); }
.slide-logo .ic { position: relative; color: #fff; }
.carousel:hover .slide { animation-play-state: paused; }
.join {
    display: inline-flex; align-items: center; gap: 8px; background: var(--gold); color: var(--navy) !important;
    text-decoration: none !important; font-weight: 700; padding: 12px 22px; border-radius: 12px; min-height: 44px;
}
.join:hover { filter: brightness(0.96); }

/* ---------- club tiles (full-colour) ---------- */
[class*="st-key-club_"] {
    border: none !important; border-radius: 22px !important; padding: 10px 10px 4px 10px;
    box-shadow: 0 12px 30px rgba(15,27,61,0.18); color: #fff;
    transition: transform .2s ease, box-shadow .2s ease;
    animation: fadein .45s ease both;
}
[class*="st-key-club_"]:hover { transform: translateY(-3px); box-shadow: 0 18px 40px rgba(15,27,61,0.28); }
@keyframes fadein { from { opacity: 0; transform: translateY(8px); } to { opacity: 1; transform: none; } }
__PER_CLUB_CSS__

.club-top { display: flex; justify-content: flex-end; margin-bottom: 8px; min-height: 28px; }
.pill {
    display: inline-flex; align-items: center; gap: 7px; background: rgba(255,255,255,0.18); border: 1px solid rgba(255,255,255,0.28);
    color: #fff; border-radius: 999px; padding: 4px 12px; font-size: 0.8rem; font-weight: 600; backdrop-filter: blur(6px);
}
.dot { width: 8px; height: 8px; border-radius: 50%; background: #4ADE80; box-shadow: 0 0 0 0 rgba(74,222,128,.7); animation: pulse 2s infinite; }
@keyframes pulse { 70% { box-shadow: 0 0 0 8px rgba(74,222,128,0); } 100% { box-shadow: 0 0 0 0 rgba(74,222,128,0); } }

/* flip card (hover / tap) */
.flip { perspective: 1000px; height: 200px; margin-bottom: 14px; }
.flip-in { position: relative; width: 100%; height: 100%; transition: transform .6s cubic-bezier(.2,.7,.2,1); transform-style: preserve-3d; }
.flip:hover .flip-in, .flip:active .flip-in { transform: rotateY(180deg); }
.face { position: absolute; inset: 0; border-radius: 16px; backface-visibility: hidden; -webkit-backface-visibility: hidden; display: flex; }
.face.front { align-items: center; justify-content: center; background: rgba(255,255,255,0.10); border: 1px solid rgba(255,255,255,0.18); overflow: hidden; }
.face.front::before { content: ""; position: absolute; width: 70%; height: 90%; border-radius: 50%; background: radial-gradient(circle, var(--glow) 0%, transparent 68%); opacity: .45; }
.face.front img { position: relative; max-width: 66%; max-height: 82%; object-fit: contain; filter: drop-shadow(0 0 22px var(--glow)) drop-shadow(0 8px 14px rgba(0,0,0,.35)); }
.face.front .ic { position: relative; color: #fff; }
.face.back { transform: rotateY(180deg); background: rgba(0,0,0,0.30); border: 1px solid rgba(255,255,255,0.2); flex-direction: column; justify-content: center; padding: 14px 20px; gap: 7px; backdrop-filter: blur(8px); }
.back-title { font-family: 'Outfit', sans-serif; font-weight: 700; font-size: 0.98rem; margin-bottom: 2px; }
.brow { display: flex; align-items: center; gap: 10px; font-size: 0.92rem; color: rgba(255,255,255,0.94); }
.brow .ic { color: var(--glow); }

.club-name { font-family: 'Outfit', sans-serif; font-weight: 700; font-size: 1.2rem; line-height: 1.2; color: #fff; }
.club-domain { font-size: 0.88rem; color: rgba(255,255,255,0.78); margin: 4px 0 10px 0; line-height: 1.35; }
.club-desc { color: rgba(255,255,255,0.9); font-size: 0.97rem; line-height: 1.55; margin: 0 0 12px 0; min-height: 3.1em; }
.tile .chip { background: rgba(255,255,255,0.16); color: #fff; border: 1px solid rgba(255,255,255,0.14); }
.chips-row { margin-bottom: 8px; min-height: 2.2em; }

[class*="st-key-club_"] .stButton > button {
    background: rgba(255,255,255,0.16); color: #fff; border: 1px solid rgba(255,255,255,0.34); backdrop-filter: blur(6px);
}
[class*="st-key-club_"] .stButton > button:hover { background: rgba(255,255,255,0.28); border-color: #fff; color: #fff; }

.badge { flex: none; display: flex; align-items: center; justify-content: center; overflow: hidden; }
.badge img { width: 100%; height: 100%; object-fit: contain; }
.badge-md { width: 52px; height: 52px; border-radius: 14px; }
.badge-lg { width: 64px; height: 64px; border-radius: 16px; }

.chip {
    display: inline-flex; align-items: center; gap: 6px; background: var(--soft); color: var(--body);
    border-radius: 9px; padding: 5px 11px; font-size: 0.82rem; font-weight: 500; margin: 0 6px 6px 0;
}

/* ---------- buttons ---------- */
.stButton > button {
    border-radius: 12px; font-weight: 600; font-family: 'DM Sans', sans-serif;
    border: 1px solid var(--line); padding: 0.55rem 1rem; min-height: 46px;
    background: var(--card); color: var(--ink);
}
.stButton > button:hover { border-color: var(--head); color: var(--head); }
.stButton > button[kind="primary"], .stButton > button[data-testid="stBaseButton-primary"] {
    background: var(--btn); color: var(--btn-ink); border: none; padding: 0.7rem 1.6rem; font-size: 1.02rem;
}
.stButton > button[kind="primary"]:hover, .stButton > button[data-testid="stBaseButton-primary"]:hover { background: var(--btn-hover); color: var(--btn-ink); }
.stButton > button[kind="primary"] p, .stButton > button[data-testid="stBaseButton-primary"] p { color: var(--btn-ink); }

/* ---------- AI panel (glass) ---------- */
.st-key-ai_panel {
    background: var(--glass); backdrop-filter: blur(18px); -webkit-backdrop-filter: blur(18px);
    border: 1px solid var(--glass-line) !important; border-radius: 24px !important; padding: 24px 28px 18px 28px;
    box-shadow: var(--shadow);
}
.st-key-quick .stButton > button { min-height: 38px; padding: 0.3rem 0.8rem; font-size: 0.88rem; border-radius: 999px; }

/* chat bubbles */
.msg { display: flex; gap: 12px; margin: 16px 0; align-items: flex-start; }
.msg.user { justify-content: flex-end; }
.avatar { width: 40px; height: 40px; border-radius: 50%; flex: none; display: flex; align-items: center; justify-content: center; background: var(--navy); color: var(--gold); box-shadow: 0 0 0 3px rgba(242,179,61,0.25); }
.bubble { border-radius: 18px; padding: 14px 18px; max-width: 78%; line-height: 1.55; }
.bubble.user { background: var(--btn); color: var(--btn-ink); border-bottom-right-radius: 6px; }
.bubble.ai { background: var(--card); border: 1px solid var(--line); border-top-left-radius: 6px; color: var(--body); width: 100%; max-width: 100%; }
.typing-line { height: 12px; border-radius: 8px; margin: 10px 0; }
.shimmer { background: linear-gradient(90deg, var(--soft) 25%, rgba(255,255,255,0.55) 50%, var(--soft) 75%); background-size: 200% 100%; animation: shimmer 1.3s linear infinite; }
@keyframes shimmer { from { background-position: 200% 0; } to { background-position: -200% 0; } }

/* ---------- recommendation result ---------- */
.result-top { display: flex; align-items: center; gap: 18px; margin-bottom: 12px; }
.result-label { color: var(--muted); font-weight: 600; font-size: 0.9rem; margin-bottom: 4px; }
.result-name { font-family: 'Outfit', sans-serif; font-size: 1.5rem; font-weight: 800; color: var(--head); line-height: 1.2; }
.ring { position: relative; width: 84px; height: 84px; flex: none; }
.ring svg { transform: rotate(-90deg); }
.ring .track { stroke: var(--soft); }
.ring .fill { stroke-linecap: round; animation: ringfill 1.4s ease-out both; }
@keyframes ringfill { from { stroke-dashoffset: 238.76; } }
.ring-num { position: absolute; inset: 0; display: flex; flex-direction: column; align-items: center; justify-content: center; font-family: 'Outfit', sans-serif; font-weight: 800; color: var(--head); font-size: 1.15rem; line-height: 1; }
.ring-num small { font-family: 'DM Sans', sans-serif; font-weight: 500; font-size: 0.64rem; color: var(--muted); margin-top: 3px; }
.glow-once { animation: glowonce 1.8s ease-out 1; }
@keyframes glowonce { 0% { box-shadow: 0 0 0 0 rgba(242,179,61,0.65); } 60% { box-shadow: 0 0 0 18px rgba(242,179,61,0); } 100% { box-shadow: 0 0 0 0 rgba(242,179,61,0); } }
.bubble p { color: var(--body); font-size: 1.03rem; margin: 0 0 14px 0; }
.result-role { background: var(--soft); border-radius: 12px; padding: 12px 16px; margin-bottom: 14px; color: var(--ink); }
.result-role b { color: var(--head); }
.bubble ul { margin: 0 0 16px 0; padding-left: 20px; color: var(--body); line-height: 1.7; }
.contact {
    display: inline-flex; align-items: center; gap: 8px; background: var(--gold); color: var(--navy) !important;
    text-decoration: none !important; font-weight: 700; padding: 11px 18px; border-radius: 12px; min-height: 44px;
}
.contact:hover { filter: brightness(0.96); }
.top3-title { font-family: 'Outfit', sans-serif; font-weight: 700; color: var(--head); margin: 20px 0 10px 0; }
.top3 { display: grid; grid-template-columns: repeat(3, 1fr); gap: 12px; }
.mini { border: 1px solid var(--line); background: var(--soft); border-radius: 14px; padding: 12px 14px; }
.mini.best { border-color: var(--gold); box-shadow: 0 0 0 2px rgba(242,179,61,0.35); background: var(--card); }
.mini-name { font-weight: 700; color: var(--head); font-size: 0.95rem; line-height: 1.25; margin: 8px 0 8px 0; }
.mini-bar { height: 7px; border-radius: 99px; background: var(--line); overflow: hidden; }
.mini-bar i { display: block; height: 100%; border-radius: 99px; animation: barfill 1.1s ease-out both; }
@keyframes barfill { from { width: 0; } }
.mini-pct { font-size: 0.8rem; color: var(--muted); margin-top: 6px; }
.mini-best-tag { font-size: 0.74rem; font-weight: 700; color: #8A5A00; background: var(--gold); border-radius: 99px; padding: 2px 9px; }

/* confetti (renders once) */
.confetti { position: fixed; inset: 0; pointer-events: none; z-index: 9999; overflow: hidden; }
.confetti i { position: absolute; top: -16px; width: 8px; height: 14px; border-radius: 2px; opacity: 0; animation: fall linear 1 forwards; }
@keyframes fall { 0% { opacity: 1; transform: translateY(0) rotate(0); } 100% { opacity: 0; transform: translateY(105vh) rotate(720deg); } }

/* ---------- compare ---------- */
.cmp-wrap { overflow-x: auto; }
.cmp { width: 100%; border-collapse: separate; border-spacing: 0; background: var(--card); border: 1px solid var(--line); border-radius: 16px; overflow: hidden; }
.cmp th, .cmp td { padding: 14px 16px; text-align: left; vertical-align: top; border-bottom: 1px solid var(--line); color: var(--body); font-size: 0.95rem; line-height: 1.5; }
.cmp th.rowh { width: 150px; color: var(--muted); font-weight: 600; background: var(--soft); }
.cmp tr:last-child td, .cmp tr:last-child th { border-bottom: none; }
.cmp thead th { color: #fff; font-family: 'Outfit', sans-serif; font-size: 1.05rem; }

/* ---------- dialog (glass) ---------- */
div[role="dialog"],
div[role="dialog"] > div,
div[role="dialog"] > div > div,
div[role="dialog"] > div > div > div,
div[data-testid="stDialog"] div[role="dialog"] [data-testid="stVerticalBlock"],
div[data-testid="stDialog"] div[role="dialog"] [data-testid="stVerticalBlockBorderWrapper"],
div[data-testid="stDialog"] div[role="dialog"] [data-testid="stElementContainer"],
div[data-testid="stDialog"] div[role="dialog"] [data-testid="stMarkdownContainer"] {
    background: var(--card) !important; background-color: var(--card) !important;
}
div[role="dialog"] { color: var(--body) !important; border: 1px solid var(--line); border-radius: 22px; overflow: hidden; }
div[role="dialog"] h1, div[role="dialog"] h2, div[role="dialog"] h3,
div[role="dialog"] [data-testid="stHeading"], div[role="dialog"] [data-testid="stHeading"] * { color: var(--head) !important; }
div[role="dialog"] p, div[role="dialog"] li, div[role="dialog"] span { color: inherit; }
div[role="dialog"] button[aria-label="Close"], div[role="dialog"] button[aria-label="Close"] svg { color: var(--head) !important; }
[class*="st-key-dlg_"] { background: transparent !important; color: var(--body); }
.dlg-head { display: flex; align-items: center; gap: 16px; margin-bottom: 14px; }
.dlg-title { font-family: 'Outfit', sans-serif; font-size: 1.5rem; font-weight: 800; color: var(--head); line-height: 1.2; }
.dlg-domain { color: var(--muted); margin-top: 4px; }
.dlg-label { font-family: 'Outfit', sans-serif; font-weight: 700; color: var(--head); font-size: 1.05rem; margin: 22px 0 8px 0; }
.dlg-list { margin: 0; padding-left: 20px; color: var(--body); line-height: 1.75; }
.callout { background: var(--soft); border-left: 4px solid var(--gold); border-radius: 10px; padding: 12px 16px; color: var(--ink); line-height: 1.55; }
.timeline { border-left: 2px solid var(--line); margin-left: 6px; padding-left: 16px; }
.tl-item { margin-bottom: 12px; color: var(--body); }
.tl-date { font-weight: 700; color: var(--head); }
.team { display: flex; flex-wrap: wrap; gap: 16px; }
.member { text-align: center; width: 92px; font-size: 0.85rem; color: var(--body); }
.member img, .member .ph { width: 72px; height: 72px; border-radius: 50%; object-fit: cover; margin: 0 auto 6px auto; background: var(--soft); display: flex; align-items: center; justify-content: center; }
.member b { display: block; color: var(--head); }
.quote { border-left: 3px solid var(--gold); padding: 4px 0 4px 14px; margin-bottom: 10px; color: var(--body); font-style: italic; }
.quote span { display: block; font-style: normal; color: var(--muted); font-size: 0.85rem; margin-top: 2px; }

/* ---------- footer ---------- */
.footer { margin-top: 46px; padding: 26px 30px; border-radius: 20px; background: var(--glass); backdrop-filter: blur(14px); border: 1px solid var(--glass-line); color: var(--muted); font-size: 0.92rem; line-height: 1.7; }
.footer b { color: var(--head); font-family: 'Outfit', sans-serif; font-size: 1.05rem; }
.footer a { color: var(--head); font-weight: 600; margin-right: 14px; }
.footer .row { display: flex; align-items: center; gap: 8px; }

/* ---------- mobile-first adjustments ---------- */
@media (max-width: 900px) { .hero-logos { display: none; } .st-key-topbar { position: static; } }
@media (max-width: 700px) {
    .hero { padding: 26px 22px; border-radius: 20px; }
    .hero-title { font-size: 2.1rem; }
    .hero-sub { font-size: 1rem; }
    .hero-top { margin-bottom: 16px; }
    .carousel { height: 340px; }
    .slide { flex-direction: column; align-items: flex-start; justify-content: center; padding: 24px 24px; gap: 8px; }
    .slide-logo { width: 84px; height: 84px; position: absolute; right: 18px; top: 18px; }
    .slide-title { font-size: 1.55rem; }
    .top3 { grid-template-columns: 1fr; }
    .bubble { max-width: 92%; }
    .result-top { flex-direction: column; align-items: flex-start; }
    .st-key-ai_panel { padding: 18px 16px 12px 16px; }
    .stButton > button { min-height: 50px; }
    .section-title { font-size: 1.4rem; }
}
</style>
"""

st.markdown(
    CSS.replace("__THEME_VARS__", theme_vars).replace("__PER_CLUB_CSS__", per_club_css),
    unsafe_allow_html=True,
)

# ---------------------------------------------------------
# Top bar (sticky): brand, navigation, dark mode
# ---------------------------------------------------------
college_logo = find_logo("dbit")
seal_inner = f'<img src="{college_logo}" alt="">' if college_logo else icon_svg("grad", 24)


def nav_control():
    if hasattr(st, "segmented_control"):
        v = st.segmented_control("Navigate", NAV_OPTIONS, default="Explore", key="nav", label_visibility="collapsed")
        return v or "Explore"
    return st.radio("Navigate", NAV_OPTIONS, horizontal=True, key="nav", label_visibility="collapsed")


with st.container(key="topbar"):
    try:
        tb1, tb2, tb3 = st.columns([3.1, 3.3, 1.1], vertical_alignment="center")
    except TypeError:
        tb1, tb2, tb3 = st.columns([3.1, 3.3, 1.1])
    with tb1:
        st.markdown(
            html_block(f"""
<div class="brand">
    <div class="brand-mark">{seal_inner}</div>
    <div>
        <div class="brand-name">{esc(COLLEGE_NAME)}</div>
        <div class="brand-sub">Campus Club Discovery</div>
    </div>
</div>
"""),
            unsafe_allow_html=True,
        )
    with tb2:
        page = nav_control()
    with tb3:
        st.toggle("Dark mode", key="dark")


def go_find():
    st.session_state["nav"] = "Find my club"


# ---------------------------------------------------------
# Hero
# ---------------------------------------------------------
def hero_logo_cluster():
    cells = []
    for i, c in enumerate(clubs_dataset):
        lg = find_logo(c["club_id"])
        inner = f'<img src="{lg}" alt="">' if lg else icon_svg(style_of(c)["icon"], 34)
        cells.append(f'<div class="lg" style="animation-delay:{-i * 0.9}s">{inner}</div>')
    return "".join(cells)


total_units = sum(len(c["sub_units"]) for c in clubs_dataset)
total_skills = len({s for c in clubs_dataset for s in c["skills_learned"]})

st.markdown(
    html_block(f"""
<div class="hero">
<div class="hero-wrap">
<div>
    <div class="hero-top">
        <div class="hero-mark">{seal_inner}</div>
        <div>
            <div class="hero-college">{esc(COLLEGE_NAME)}</div>
            <div class="hero-tag">Learn it. Build it. Belong.</div>
        </div>
    </div>
    <div class="hero-title">Find the club where you belong</div>
    <div class="hero-sub">Browse every student club on campus, or tell us what you enjoy and we will point you to the one that fits.</div>
    <div class="hero-stats">
        <div class="hero-stat"><b>{len(clubs_dataset)}</b>clubs</div>
        <div class="hero-stat"><b>{total_units}</b>teams you can join</div>
        <div class="hero-stat"><b>{total_skills}</b>skills you can pick up</div>
    </div>
</div>
<div class="hero-logos">{hero_logo_cluster()}</div>
</div>
</div>
"""),
    unsafe_allow_html=True,
)

if page == "Explore":
    st.button("Find my club", type="primary", on_click=go_find, key="hero_cta")


# ---------------------------------------------------------
# Club details (popup, with a fallback for older Streamlit)
# ---------------------------------------------------------
def render_club_details(club):
    with st.container(key=f"dlg_{club['club_id']}"):
        _render_club_details_inner(club)


def _render_club_details_inner(club):
    s = style_of(club)
    cid = club["club_id"]
    roles = club.get("roles_open")
    extras = CLUB_EXTRAS.get(cid, {})

    events_html = ""
    if extras.get("events"):
        items = "".join(
            f'<div class="tl-item"><span class="tl-date">{esc(e.get("date", ""))}</span> &nbsp; {esc(e.get("title", ""))}</div>'
            for e in extras["events"][:3]
        )
        events_html = f'<div class="dlg-label">Upcoming events</div><div class="timeline">{items}</div>'

    team_html = ""
    if extras.get("team"):
        members = []
        for m in extras["team"]:
            photo = m.get("photo")
            if photo and os.path.exists(photo):
                pic = f'<img src="{data_uri(photo)}" alt="">'
            else:
                pic = f'<div class="ph">{icon_svg("users", 26)}</div>'
            members.append(f'<div class="member">{pic}<b>{esc(m.get("name", ""))}</b>{esc(m.get("role", ""))}</div>')
        team_html = f'<div class="dlg-label">Meet the team</div><div class="team">{"".join(members)}</div>'

    quotes_html = ""
    if extras.get("quotes"):
        quotes_html = '<div class="dlg-label">What members say</div>' + "".join(
            f'<div class="quote">“{esc(q.get("text", ""))}”<span>{esc(q.get("by", ""))}</span></div>' for q in extras["quotes"][:3]
        )

    phone = icon_svg("phone", 16)
    st.markdown(
        html_block(f"""
<div class="dlg-head">
    {badge_html(club, "lg")}
    <div>
        <div class="dlg-title">{esc(club["club_name"])}</div>
        <div class="dlg-domain">{esc(club["domain"])}</div>
    </div>
</div>
<p style="color:var(--body);line-height:1.6;margin:0;">{esc(club["description"])}</p>
<div class="dlg-label">Teams you can join</div>
<div>{chips_html(club["sub_units"], s["solid"])}</div>
<div class="dlg-label">What the club does</div>
<ul class="dlg-list">{"".join(f"<li>{esc(a)}</li>" for a in club["key_activities"])}</ul>
<div class="dlg-label">Skills you will pick up</div>
<div>{chips_html(club["skills_learned"], icons=True)}</div>
{f'<div class="dlg-label">Open roles</div><div>{chips_html(roles)}</div>' if roles else ""}
<div class="dlg-label">How to join</div>
<div class="callout">{esc(club["membership_requirements"])}</div>
{events_html}
{team_html}
{quotes_html}
"""),
        unsafe_allow_html=True,
    )

    gallery = os.path.join(GALLERY_DIR, cid)
    if os.path.isdir(gallery):
        pics = sorted(
            os.path.join(gallery, f) for f in os.listdir(gallery)
            if f.lower().endswith((".png", ".jpg", ".jpeg", ".webp"))
        )
        if pics:
            st.markdown('<div class="dlg-label">From past events</div>', unsafe_allow_html=True)
            st.image(pics[:6], width=150)

    if extras.get("video"):
        st.markdown('<div class="dlg-label">Watch</div>', unsafe_allow_html=True)
        st.video(extras["video"])

    st.markdown(
        f'<div style="margin-top:20px;"><a class="contact" href="tel:{esc(club["contact"])}">{phone}Call {esc(club["contact"])}</a></div>',
        unsafe_allow_html=True,
    )


HAS_DIALOG = hasattr(st, "dialog")

if HAS_DIALOG:
    @st.dialog("Club details", width="large")
    def club_dialog(club):
        render_club_details(club)


# ---------------------------------------------------------
# Explore: carousel, filters, tiles
# ---------------------------------------------------------
def carousel_html():
    n = len(clubs_dataset)
    slot = 100.0 / n
    dur = n * 6
    keyframes = (
        f"@keyframes slideshow {{0%{{opacity:0;visibility:visible}} {slot*0.12:.2f}%{{opacity:1;visibility:visible}} "
        f"{slot*0.88:.2f}%{{opacity:1;visibility:visible}} {slot:.2f}%{{opacity:0;visibility:hidden}} "
        f"100%{{opacity:0;visibility:hidden}}}}"
    )
    slides = []
    for i, c in enumerate(clubs_dataset):
        s = style_of(c)
        logo = find_logo(c["club_id"])
        art = f'<img src="{logo}" alt="">' if logo else icon_svg(s["icon"], 96, 1.5)
        slides.append(f"""
<div class="slide" style="background:linear-gradient(135deg,{s["from"]},{s["to"]});--g:{s["glow"]};animation:slideshow {dur}s linear {i * 6}s infinite;">
    <div class="slide-text">
        <div class="slide-kicker">{icon_svg("spark", 15)} Featured club</div>
        <div class="slide-title">{esc(c["club_name"])}</div>
        <div class="slide-pitch">{esc(short_text(c["description"], 110))}</div>
        <a class="join" href="tel:{esc(c["contact"])}">{icon_svg("phone", 16)}Join</a>
    </div>
    <div class="slide-logo">{art}</div>
</div>""")
    return html_block(f"<style>{keyframes}</style><div class=\"carousel\">{''.join(slides)}</div>")


def tile_html(club, wide):
    s = style_of(club)
    logo = find_logo(club["club_id"])
    front = f'<img src="{logo}" alt="">' if logo else icon_svg(s["icon"], 72, 1.5)
    back_rows = "".join(
        f'<div class="brow">{icon_svg(skill_icon(k), 15)}<span>{esc(k)}</span></div>'
        for k in club["skills_learned"][:5]
    )
    return html_block(f"""
<div class="tile">
<div class="club-top"><span class="pill"><i class="dot"></i>{esc(club_status(club))}</span></div>
<div class="flip"><div class="flip-in">
    <div class="face front">{front}</div>
    <div class="face back"><div class="back-title">Top skills</div>{back_rows}</div>
</div></div>
<div class="club-name">{esc(club["club_name"])}</div>
<div class="club-domain">{esc(club["domain"])}</div>
<p class="club-desc">{esc(short_text(club["description"], 150 if wide else 118))}</p>
<div class="chips-row">{chips_html(club["skills_learned"], limit=5 if wide else 3, icons=True)}</div>
</div>
""")


def choose_category():
    options = ["All"] + list(CATEGORIES.keys())
    if hasattr(st, "pills"):
        v = st.pills("Category", options, default="All", key="cat", label_visibility="collapsed")
        return v or "All"
    return st.radio("Category", options, horizontal=True, key="cat", label_visibility="collapsed")


def render_tile(club, wide):
    with st.container(border=True, key=f"club_{club['club_id']}"):
        st.markdown(tile_html(club, wide), unsafe_allow_html=True)
        if HAS_DIALOG:
            if st.button("View details", key=f"btn_{club['club_id']}", use_container_width=True):
                club_dialog(club)
        else:
            with st.expander("View details"):
                render_club_details(club)


def render_explore():
    st.markdown(carousel_html(), unsafe_allow_html=True)

    st.markdown('<div class="section-title">Explore campus clubs</div>', unsafe_allow_html=True)
    st.markdown(
        '<div class="section-sub">Filter by category or search a skill. Hover a tile to see its top skills, then open it for the full details.</div>',
        unsafe_allow_html=True,
    )

    category = choose_category()
    query = st.text_input(
        "Search clubs",
        placeholder="Search by club, skill or activity, e.g. AWS, dance, research",
        key="club_search",
        label_visibility="collapsed",
    )

    visible = filter_clubs(category, query)
    if not visible:
        st.info("No club matches that search. Clear the filter or try a different word.")
        return

    # Alternating wide / narrow rows give the grid some rhythm
    patterns = [(2, 1), (1, 2), (1, 1)]
    row = 0
    for i in range(0, len(visible), 2):
        chunk = visible[i:i + 2]
        pat = patterns[row % len(patterns)]
        row += 1
        if len(chunk) == 1:
            cols = st.columns([2, 1], gap="medium")
            with cols[0]:
                render_tile(chunk[0], wide=True)
        else:
            cols = st.columns(list(pat), gap="medium")
            for col, club, w in zip(cols, chunk, pat):
                with col:
                    render_tile(club, wide=(w == 2))


# ---------------------------------------------------------
# Core AI Recommendation Logic (unchanged)
# ---------------------------------------------------------
def generate_club_recommendation(student_query, dataset):
    clubs_text = json.dumps(dataset, indent=2)

    prompt = f"""
You are an expert university club advisor.
Analyze the student input and recommend the single best matching club from the dataset.

STUDENT INPUT / INTERESTS:
"{student_query}"

CLUBS DATASET:
{clubs_text}

Return ONLY valid JSON with this exact structure:

{{
    "matched_club_name": "",
    "match_reason": "",
    "suggested_role_or_subunit": "",
    "key_benefits": [],
    "contact": ""
}}
"""

    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.3
        )

        text = response.choices[0].message.content
        return json.loads(text)
    except Exception as e:
        return {"error": str(e)}


# ---------------------------------------------------------
# Extra call used only for the match meter and top-3 cards.
# It never changes which club is recommended above.
# ---------------------------------------------------------
def generate_match_scores(student_query, dataset):
    names = [c["club_name"] for c in dataset]
    prompt = f"""
Rate how well each club fits this student, from 0 to 100.

STUDENT INPUT / INTERESTS:
"{student_query}"

CLUBS DATASET:
{json.dumps(dataset, indent=2)}

Return ONLY valid JSON with this exact structure, using these exact club names {json.dumps(names)}:

{{
    "scores": [{{"club_name": "", "percent": 0}}]
}}
"""
    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.2
        )
        text = response.choices[0].message.content.strip()
        text = text.replace("```json", "").replace("```", "").strip()
        data = json.loads(text)
        scores = {}
        for item in data.get("scores", []):
            c = club_by_name(item.get("club_name", ""))
            if c:
                scores[c["club_id"]] = max(0, min(100, int(round(float(item.get("percent", 0))))))
        return scores
    except Exception:
        return {}


# ---------------------------------------------------------
# AI panel pieces
# ---------------------------------------------------------
AVATAR = f'<div class="avatar">{icon_svg("spark", 20)}</div>'


def user_bubble(text):
    return f'<div class="msg user"><div class="bubble user">{esc(text)}</div></div>'


def skeleton_html(query):
    return html_block(f"""
{user_bubble(query)}
<div class="msg">{AVATAR}<div class="bubble ai">
<div class="typing-line shimmer" style="width:42%"></div>
<div class="typing-line shimmer" style="width:92%"></div>
<div class="typing-line shimmer" style="width:78%"></div>
<div class="typing-line shimmer" style="width:55%"></div>
</div></div>
""")


def confetti_html():
    colors = ["#F2B33D", "#8B3FE0", "#0F7BC0", "#E8710A", "#4ADE80", "#FF6B6B"]
    pieces = []
    for _ in range(36):
        pieces.append(
            f'<i style="left:{random.uniform(2, 98):.1f}%;background:{random.choice(colors)};'
            f'animation-delay:{random.uniform(0, 0.7):.2f}s;animation-duration:{random.uniform(2.0, 3.4):.2f}s;"></i>'
        )
    return f'<div class="confetti">{"".join(pieces)}</div>'


def ring_html(percent, color):
    c = 238.76
    offset = c * (1 - percent / 100)
    return html_block(f"""
<div class="ring">
<svg width="84" height="84" viewBox="0 0 90 90">
<circle class="track" cx="45" cy="45" r="38" fill="none" stroke-width="9"/>
<circle class="fill" cx="45" cy="45" r="38" fill="none" stroke="{color}" stroke-width="9" stroke-dasharray="{c}" stroke-dashoffset="{offset:.2f}"/>
</svg>
<div class="ring-num">{percent}%<small>match</small></div>
</div>
""")


def top3_html(matched_club, scores):
    if not scores:
        return ""
    order = sorted(scores.items(), key=lambda kv: kv[1], reverse=True)
    ids = [matched_club["club_id"]] if matched_club else []
    ids += [cid for cid, _ in order if cid not in ids]
    ids = ids[:3]
    cards = []
    for idx, cid in enumerate(ids):
        club = next((c for c in clubs_dataset if c["club_id"] == cid), None)
        if not club or cid not in scores:
            continue
        s = style_of(club)
        pct = scores[cid]
        best = idx == 0
        tag = '<span class="mini-best-tag">Best match</span>' if best else ""
        cards.append(f"""
<div class="mini {'best' if best else ''}">
<div style="display:flex;align-items:center;justify-content:space-between;gap:8px;">{badge_html(club)}{tag}</div>
<div class="mini-name">{esc(club["club_name"])}</div>
<div class="mini-bar"><i style="width:{pct}%;background:linear-gradient(90deg,{s["from"]},{s["solid"]});"></i></div>
<div class="mini-pct">{pct}% match</div>
</div>""")
    if not cards:
        return ""
    return f'<div class="top3-title">Your top matches</div><div class="top3">{"".join(cards)}</div>'


def render_result(last):
    query = last["query"]
    recommendation = last["rec"]

    if "error" in recommendation:
        st.markdown(user_bubble(query), unsafe_allow_html=True)
        st.error(f"Could not get a recommendation. Details: {recommendation['error']}")
        return

    matched_name = recommendation.get("matched_club_name", "")
    matched_club = club_by_name(matched_name)
    ms = style_by_name(matched_name)
    scores = last.get("scores") or {}
    percent = scores.get(matched_club["club_id"]) if matched_club else None

    badge = badge_html(matched_club) if matched_club else f'<div class="badge badge-md" style="background:{tint(ms["solid"])};color:{ms["solid"]};">{icon_svg(ms["icon"], 26)}</div>'
    ring = ring_html(percent, ms["solid"]) if percent is not None else ""
    benefits = "".join(f"<li>{esc(b)}</li>" for b in recommendation.get("key_benefits", []))
    contact = str(recommendation.get("contact", ""))
    glow = "glow-once" if last.get("fresh") else ""

    if last.get("fresh"):
        st.markdown(confetti_html(), unsafe_allow_html=True)
        last["fresh"] = False

    st.markdown(
        html_block(f"""
{user_bubble(query)}
<div class="msg">{AVATAR}
<div class="bubble ai {glow}" style="border-left:5px solid {ms["solid"]};">
    <div class="result-top">
        {ring}
        {badge}
        <div>
            <div class="result-label">Your best match</div>
            <div class="result-name">{esc(matched_name)}</div>
        </div>
    </div>
    <p>{esc(recommendation.get("match_reason", ""))}</p>
    <div class="result-role"><b>Where to start:</b> {esc(recommendation.get("suggested_role_or_subunit", ""))}</div>
    {f"<ul>{benefits}</ul>" if benefits else ""}
    {f'<a class="contact" href="tel:{esc(contact)}">{icon_svg("phone", 16)}Call {esc(contact)}</a>' if contact else ""}
    {top3_html(matched_club, scores)}
</div>
</div>
"""),
        unsafe_allow_html=True,
    )


def set_interest(text):
    st.session_state["interest"] = text


def render_find():
    st.markdown('<div class="section-title" style="margin-top:8px;">Not sure where you fit?</div>', unsafe_allow_html=True)
    st.markdown(
        '<div class="section-sub">Describe what you enjoy or want to learn. We will match you with a club and suggest where to start.</div>',
        unsafe_allow_html=True,
    )

    with st.container(key="ai_panel"):
        last = st.session_state.get("last")
        if not last:
            st.markdown(
                html_block(f"""
<div class="msg">{AVATAR}<div class="bubble ai">Hi! Tell me what you enjoy or what you want to learn, and I will find the club that fits. You can also tap one of the quick starts below.</div></div>
"""),
                unsafe_allow_html=True,
            )

        with st.container(key="quick"):
            qcols = st.columns(len(QUICK_STARTS))
            for col, text in zip(qcols, QUICK_STARTS):
                with col:
                    st.button(text, key=f"qs_{text}", on_click=set_interest, args=(text,), use_container_width=True)

        student_interest_input = st.text_area(
            "What are your interests or what do you want to learn?",
            placeholder="e.g., I want to learn cloud architecture, practice AWS deployment, and help with tech operations.",
            height=120,
            key="interest",
        )

        find_clicked = st.button("Find my club", type="primary", key="find_btn")

        if find_clicked:
            if student_interest_input.strip():
                ph = st.empty()
                ph.markdown(skeleton_html(student_interest_input), unsafe_allow_html=True)
                recommendation = generate_club_recommendation(student_interest_input, clubs_dataset)
                scores = {}
                if "error" not in recommendation:
                    scores = generate_match_scores(student_interest_input, clubs_dataset)
                ph.empty()
                st.session_state["last"] = {
                    "query": student_interest_input,
                    "rec": recommendation,
                    "scores": scores,
                    "fresh": True,
                }
            else:
                st.warning("Tell us a little about your interests first.")

        last = st.session_state.get("last")
        if last:
            render_result(last)


# ---------------------------------------------------------
# Compare
# ---------------------------------------------------------
def render_compare():
    st.markdown('<div class="section-title" style="margin-top:8px;">Compare clubs</div>', unsafe_allow_html=True)
    st.markdown(
        '<div class="section-sub">Pick two or three clubs to see them side by side.</div>',
        unsafe_allow_html=True,
    )
    names = [c["club_name"] for c in clubs_dataset]
    picked = st.multiselect("Clubs to compare", names, default=names[:2], max_selections=3, label_visibility="collapsed")
    chosen = [club_by_name(n) for n in picked]
    chosen = [c for c in chosen if c]
    if len(chosen) < 2:
        st.info("Select at least two clubs to compare.")
        return

    head = "".join(
        f'<th style="background:linear-gradient(135deg,{style_of(c)["from"]},{style_of(c)["to"]});">{esc(c["club_name"])}</th>'
        for c in chosen
    )

    def row(label, fn):
        cells = "".join(f"<td>{fn(c)}</td>" for c in chosen)
        return f'<tr><th class="rowh">{label}</th>{cells}</tr>'

    rows = [
        row("Focus", lambda c: esc(c["domain"])),
        row("Teams", lambda c: chips_html(c["sub_units"], style_of(c)["solid"])),
        row("Skills", lambda c: chips_html(c["skills_learned"], icons=True)),
        row("Activities", lambda c: "<br>".join(esc(a) for a in c["key_activities"])),
        row("Open roles", lambda c: chips_html(c["roles_open"]) if c.get("roles_open") else "Not listed"),
        row("How to join", lambda c: esc(c["membership_requirements"])),
        row("Contact", lambda c: f'<a class="contact" href="tel:{esc(c["contact"])}">{icon_svg("phone", 15)}{esc(c["contact"])}</a>'),
    ]
    st.markdown(
        html_block(f'<div class="cmp-wrap"><table class="cmp"><thead><tr><th class="rowh" style="background:var(--soft);"></th>{head}</tr></thead><tbody>{"".join(rows)}</tbody></table></div>'),
        unsafe_allow_html=True,
    )


# ---------------------------------------------------------
# Page router
# ---------------------------------------------------------
if page == "Explore":
    render_explore()
elif page == "Find my club":
    render_find()
else:
    render_compare()

# ---------------------------------------------------------
# Footer
# ---------------------------------------------------------
footer_lines = [f"<b>{esc(COLLEGE_NAME)}</b>"]
if COLLEGE_ADDRESS:
    footer_lines.append(f'<div class="row">{icon_svg("pin", 15)}{esc(COLLEGE_ADDRESS)}</div>')
if COORDINATOR_PHONE:
    who = f" ({esc(COORDINATOR_NAME)})" if COORDINATOR_NAME else ""
    footer_lines.append(
        f'<div class="row">{icon_svg("phone", 15)}Contact the coordinator{who}: <a href="tel:{esc(COORDINATOR_PHONE)}" style="margin-left:4px;">{esc(COORDINATOR_PHONE)}</a></div>'
    )
if SOCIAL_LINKS:
    links = "".join(f'<a href="{esc(url)}" target="_blank" rel="noopener">{esc(name)}</a>' for name, url in SOCIAL_LINKS.items())
    footer_lines.append(f"<div>{links}</div>")
footer_lines.append("Campus Club Discovery Assistant")

st.markdown(f'<div class="footer">{"".join(l if l.startswith("<div") else f"<div>{l}</div>" for l in footer_lines)}</div>', unsafe_allow_html=True)
````
